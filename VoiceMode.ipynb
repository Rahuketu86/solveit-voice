So you actually have **four local components** to stand up (and one to glue):

| # | Component | Candidates | Lives where |
|---|---|---|---|

| 3 | **TTS server** | VibeVoice, Kokoro, Piper, XTTS-v2 | Python service, HTTP endpoint |
| 4 | **Extension code** | Modified `voice.js` + new provider implementations | The existing repo |
| 5 | **Tool registry** | Whatever functions the agent can call | Probably Python, alongside the LLM |

## The three real design choices

Before picking models, three decisions shape everything else:

**(a) Where do the servers run?** Same machine as the browser (localhost), or a separate box? This decides whether models fit (GPU VRAM) and whether you need HTTPS/auth on the endpoints.

**(b) Does the local LLM *replace* Solveit's AI, or *sit beside* it?** Two very different architectures:

- **Replace**: voice → local LLM → tools → spoken reply. Solveit is bypassed entirely, the extension becomes its own little agent.
- **Beside**: voice → local LLM decides {handle locally with a tool} or {forward to Solveit}. Solveit still does the heavy lifting for coding questions; local tools handle "set a timer", "what's on my calendar", "search my notes".

**(c) What tools matter?** "Tool calling" is abstract until you name 3–5 concrete tools you want. The tool list usually decides the LLM size (a 3B model can route to a few obvious tools; reliable JSON-schema tool calling at the level of Claude/GPT-4 needs 8B+ and good prompting).

## Model picks for a T4 (16 GB) / T10 (24 GB)

Sized so all three fit on one GPU if you want (you can also split them across containers):

| Slot | T4 (16 GB) pick | T10 (24 GB) pick | VRAM |
|---|---|---|---|
| LLM with tool-calling | Qwen2.5-7B-Instruct (Q4_K_M via Ollama) | Qwen2.5-14B-Instruct (Q4) | 5–10 GB |
| STT | faster-whisper large-v3 | Whisper large-v3 / Voxtral-Mini-3B | 3–6 GB |
| TTS | Kokoro-82M (300 MB, fast, very natural) or Piper (CPU) | VibeVoice or XTTS-v2 | 0.3–4 GB |

**Why Qwen2.5 for the LLM**: native tool-calling via Hermes-style schema, works well via Ollama's OpenAI-compatible API, 7B is the smallest that's reliably good at structured tool calls. Llama 3.1 8B is the alternative.

**Voxtral note**: Voxtral is interesting because it's audio-in **and** does instruction-following — you could collapse STT + LLM into one model. But it's 3B/24B and the 3B is just-OK at tools. Worth knowing about but I'd not start there.

## The screenshot tool — one design question

CDP screenshots have two flavors and they're different beasts:

- **(a) Capture the user's current tab.** Easy: the Chrome extension's `chrome.tabs.captureVisibleTab()` API gives you a PNG of the visible tab with no CDP at all. The LLM-side "tool" just returns a marker; the extension intercepts and fills in the image.
- **(b) "Open a URL, take a screenshot, return the image."** Needs a headless Chrome in a docker container (Playwright/Puppeteer). Heavier, but more agentic — the LLM can browse without your tab.

You've got `ComputerUseDemo.ipynb` and `CDPDemo/` in the home dir, so you've been here before. Which flavor were you picturing?

## Suggested build sequence

There's an order that minimises debugging pain. I'd argue **text-first, voice-on-top**:

1. **Stand up local LLM** in docker (`ollama serve` is the 30-minute version), expose via cloudflare tunnel. *Validate: `curl /v1/chat/completions` works.*
2. **Write `local_agent.py` actor** in memexsolve. Wire it to the LLM, register existing MCP tools. *Validate: type into a memexsolve dialog, get tool-calling responses back. Zero voice involved.*
3. **Add the two new tools** (`take_screenshot`, `get_weather`). *Validate: ask the agent in text.*
4. **STT service** in docker (faster-whisper behind FastAPI). *Validate: curl audio, get text.*
5. **TTS service** in docker. *Validate: curl text, get audio.*
6. **Adapt the browser extension** — point it at memexsolve, swap STT/TTS providers to your local endpoints.

The reason for this order: every voice problem ("did it not work because of the mic, the network, the STT, the LLM, or the TTS?") collapses to "is the text version working?" If yes, the bug's in audio plumbing; if no, the bug's in the agent. You'll save days.

## The single most important up-front decision

**Should the services mimic OpenAI's audio API, or be custom?**

The existing `solveit-voice` extension already talks to `api.openai.com/v1/audio/speech` for TTS using streaming `audio/mpeg`. If your local TTS exposes the **same request/response shape**, the entire `_speakOpenAI()` method in `voice.js` works unchanged — you just swap the base URL. Same logic for STT: if you implement Whisper's `/v1/audio/transcriptions` shape, you can use any OpenAI-style client off the shelf, and the actor code in memexsolve has fewer custom abstractions to maintain.

OpenAI's audio endpoints (paraphrased):

```
POST /v1/audio/transcriptions      (STT)
  multipart/form-data:
    file:           <audio bytes>
    model:          whisper-1
    response_format: json | text | verbose_json
  → 200  {"text": "..."}              (or richer if verbose_json)

POST /v1/audio/speech                (TTS)
  application/json:
    {model, voice, input, response_format: "mp3"|"opus"|"wav"|"pcm"}
  → 200  audio/mpeg  (chunked transfer encoding — streamable)
```

**My recommendation: be OpenAI-compatible by default.** Two reasons: (a) reusing the existing extension code is a real win, and (b) you'll find an ecosystem of SDKs that just work. Only diverge when you have a concrete need (e.g. word-timestamps, voice-cloning input, multi-speaker, custom auth).

## Design dimensions to nail down

For **STT**:

| Dimension | Default | Why |
|---|---|---|
| Audio in | WebM/Opus (browser default from MediaRecorder) | No re-encoding client-side; server decodes via ffmpeg |
| Batch vs streaming | **Batch** | Voice commands are short (<10 s); streaming Whisper is fiddly and adds little |
| Response | `{text}` for v1; add `language, duration_ms, segments` if you find a need | Keep small, evolve later |
| Concurrency | Single-process, async queue if needed | T4 will OOM if two large transcriptions run in parallel; queue or 503 |
| Auth | Bearer token (shared secret) | Cloudflare tunnel = publicly reachable; you must guard it |

For **TTS**:

| Dimension | Default | Why |
|---|---|---|
| Output format | `audio/mpeg` chunked | Already streamed-consumed by `voice.js` via MediaSource |
| **Streaming** | **Yes — chunked HTTP transfer** | Time-to-first-audio (TTFA) is the difference between "fast" and "feels slow". Synthesize sentence-by-sentence, flush as you go. |
| Text preprocessing | Strip code blocks, emoji, markdown link syntax | Already done in `voice.js` `_extractProse`; could move server-side or keep client-side |
| Voice selection | Pass through to Kokoro (`af_bella`, `am_adam`, …) | Same `voice` field as OpenAI uses |
| Max input length | ~2000 chars per request | Kokoro slows on long inputs; split sentences if longer |

## The one trade-off worth pausing on: TTS streaming

This is the only decision where I'd actively push back on a "just do it simply" instinct. Two reasons it matters:

- **Latency budget**: synthesising a 10-second response can take 3–5 s on Kokoro/T4. If you wait for the full audio before sending bytes, the user hears nothing for those 5 s — feels broken. Streaming chunk-by-chunk, the first audio arrives in ~500 ms.
- **Existing client code expects it**: `voice.js`'s `_speakOpenAI` already uses `MediaSource` + `SourceBuffer.appendBuffer()` to play streamed mpeg. Match that and you inherit working code.

The "simple" version is: synthesise full audio → write whole response → done. The "right" version: synthesise sentence-by-sentence, write each chunk to `response`, flush, repeat. Kokoro is fast enough that this works.

## Other things worth deciding before you write code

- **Are STT and TTS one container or two?** Two is cleaner (independent scaling, simpler deps), one is fewer moving parts. I'd say two.
- **Health checks** — `GET /health` returns `{model_loaded: true, vram_mb: 5400}`. Helps debugging when something's wedged.
- **Logging** — every request logs `{duration_ms, input_size, output_size}`. Cheap and invaluable.
- **Idempotency / caching** — TTS: identical `(text, voice)` → identical audio. Worth a content-hash cache (disk or in-memory LRU) for re-played AI responses.

Good — that's a great learning moment. Let me unpack STT streaming vs batch properly, because the trade-off is more interesting than it looks.

## What "streaming STT" actually means

**Batch**:
```
[User speaks 3 s] [silence timer 1.5 s] [upload 200 ms] [Whisper inference 500 ms] → text
                                                                                ↑
                                                                user sees text ~2.2 s after stopping
```

**Streaming**:
```
[User speaks 3 s, audio chunks every 100 ms]
       ↑                ↑                 ↑
   "play"            "play a"         "play a sine wave"  ← interim transcripts arrive as you talk

[user stops]  [final transcript ready ~100 ms later]
```

The difference is: in batch the work happens *after* the user stops; in streaming the work happens *during* speech, so when they stop, you're nearly done.

## Why this is genuinely harder than it sounds for Whisper

Whisper (and Voxtral) was trained on **30-second clips** with full bidirectional attention. It's not architecturally designed for streaming. To stream it you have to do one of these uncomfortable things:

| Approach | Problem |
|---|---|
| Re-run on each new chunk (1 s, 2 s, 3 s of audio) | Whisper hallucinates badly on <2 s clips. Quadratic compute. |
| Sliding 30 s window | Same model run every chunk → expensive on T4 |
| VAD + segment | Lose "live" feedback; basically batch with smaller batches |
| Use a model *designed* for streaming (Parakeet, NeMo Conformer-CTC, RNN-T) | Different model entirely. Lower quality, but actually streaming-native. |

The OpenAI Whisper API itself is **batch-only** for this reason — they don't even offer streaming. So if you stay OpenAI-compatible, batch is the natural fit.

## What actually moves your latency

Look at where the seconds go in batch:

```
[silence timer 1500 ms]  ←─── this is the killer, NOT inference
[upload 200 ms]
[Whisper inference 400-800 ms on T4]
─────────────────────────
~2.2 s perceived wait
```

The silence timer is **far** bigger than inference. So the real latency lever isn't "stream the STT" — it's **replace the dumb timer with VAD**:

- **Voice Activity Detection** = a tiny model (silero-vad is ~2 MB, runs in-browser or server-side) that detects when speech actually ends, not when arbitrary silence elapses.
- Result: end-of-speech detected ~150–300 ms after user stops (vs. 1500 ms).
- Combined with batch Whisper: total perceived latency drops to ~600–800 ms — well under your TTS target.

## What I'd recommend

**Batch STT + VAD on the client side.** Reasons:

1. Stays OpenAI-compatible (`POST /v1/audio/transcriptions`, multipart, returns `{text}`).
2. Whisper/Voxtral are batch-native; no fighting the model.
3. VAD attacks the actual latency bottleneck (the silence timer).
4. Less code: one HTTP POST after speech ends, vs. a WebSocket protocol.
5. You keep streaming-feel for the *user* if you also wire Chrome's `SpeechRecognition` for purely-visual interim transcripts (no extra server work).

The only reason I'd reach for true streaming STT is **long-form dictation** (multi-minute monologues). For voice commands and short questions, batch + VAD is the sweet spot and is what most production voice assistants actually use.

## So the final shape of the STT service

```
POST /v1/audio/transcriptions
  Authorization: Bearer <token>
  multipart/form-data:
    file:            <webm/opus blob from MediaRecorder>
    model:           whisper-1   (ignored, you only have one)
    response_format: json
  →  200  {"text": "...", "duration_ms": 412}
  →  401  unauthorized
  →  415  unsupported audio format
  →  429  busy (queue full)
  →  503  model not loaded yet

GET /health  →  {"model_loaded": true, "vram_mb": 1500, "queue_depth": 0}
```

VAD lives in the **extension**, not the server — it's a client-side concern (you don't want to ship raw mic audio across the tunnel for the server to figure out when you stopped).

### Internal pipeline

```
input text
   │
   ▼
sentence splitter (regex on ". ! ?")
   │
   ▼  one sentence at a time
Kokoro pipeline  → PCM float32 chunks (24 kHz mono)
   │
   ▼
lameenc encoder (single instance, fed sequentially)
   │
   ▼  MP3 byte chunks
yield to StreamingResponse → browser
```

The reason for sentence-by-sentence: Kokoro synthesises a full sentence in one shot, but starts emitting in ~150–300 ms. If you feed it the whole 5-sentence response, you wait for all 5. If you split first and feed one at a time, the user hears sentence 1 before sentences 2–5 are even generated. That's the 500-ms TTFA.

The reason for a **single persistent lameenc instance** (not per-sentence): an MP3 stream is a sequence of self-contained frames, but a fresh encoder may add a leading silence/padding. One encoder fed continuous PCM = one clean MP3 stream.

2. **MP3 encoder library**:

   - `pyav` (ffmpeg bindings) — heavier, but more formats if you ever want Opus.
   - Recommend `lameenc` for v1.

3. **Concurrency model**:
   - v1: single asyncio lock — one synthesis at a time, others wait. Simple, safe for one user.
   - v2: per-user queue with backpressure.
   - Recommend lock for now.

4. **Voice catalog**:

Kokoro is **82M parameters total** — that's tiny, especially split across 9 languages. The model was trained primarily on English; non-English languages got less data:

- **Hindi (h)**: works, but noticeably less natural than English. Pronunciation of English loanwords in Hindi sentences (code-mixing — very common in real Indian usage) is hit-or-miss. Numbers and dates can be awkward.
- **Spanish, French, Italian, Portuguese**: similar — passable but not flagship.

If your users will speak/hear pure Hindi or pure English: Kokoro is fine.

|---|---|---|---|

## Base image: 3 realistic choices

| Base | Size | Pros | Cons |
|---|---|---|---|
| `nvidia/cuda:12.1-runtime-ubuntu22.04` | ~2.5 GB | Minimal | You install PyTorch yourself — version-matching pain |
| **`pytorch/pytorch:2.4.0-cuda12.1-cudnn9-runtime`** | ~5.5 GB | PyTorch + CUDA + cuDNN pre-installed, version-matched | Bigger image |
| `nvcr.io/nvidia/pytorch:24.05-py3` | ~15 GB | NVIDIA's optimized stack | Overkill for inference, huge |

**Pick the middle one.** The size hit is worth not fighting CUDA/PyTorch version mismatches. The exact PyTorch tag depends on what's current — `pytorch/pytorch:2.4.0-cuda12.1-cudnn9-runtime` is a safe pick as of now; check Docker Hub for the latest stable.

Why `-runtime` and not `-devel`: `-devel` includes nvcc, headers, and build tooling (~3 GB extra) that you don't need to *run* an already-built PyTorch model.

## System dependencies inside the image

Only one that matters for Kokoro:

- **`espeak-ng`** — phonemizer. Without it: cryptic `RuntimeError: Failed to find espeak-ng`. Apt-installable.

`ffmpeg` is **not** needed if you're using `lameenc` (LAME is statically bundled). Save the ~100 MB.

`build-essential` is **not** needed — `lameenc` ships wheels for common platforms.

## Host-side requirements (often the real blocker)

For `--gpus all` to do anything, the host needs:

1. **NVIDIA driver** (compatible with the CUDA version in your image — driver 535+ for CUDA 12.x).

## Dockerfile shape

The skeleton (I'm showing structure, you fill in versions/details):

```dockerfile
FROM pytorch/pytorch:2.4.0-cuda12.1-cudnn9-runtime

RUN apt-get update && apt-get install -y --no-install-recommends \
        espeak-ng \
    && rm -rf /var/lib/apt/lists/*

WORKDIR /app
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

COPY main.py .

# HF model cache → bind a host volume so weights survive container restarts
ENV HF_HOME=/cache/hf

EXPOSE 8001
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8001"]
```

The five non-obvious choices:

1. **`--no-install-recommends`** — keeps apt from pulling 200 MB of optional packages.
2. **`rm -rf /var/lib/apt/lists/*`** — drops apt cache from the image layer.
3. **`--no-cache-dir`** for pip — same idea, no wheel cache lingering.
4. **`HF_HOME=/cache/hf`** — Hugging Face's default is `~/.cache/huggingface`, which dies with the container. Mount a volume here so the 330 MB Kokoro download survives restarts.
5. **`COPY requirements.txt` separately before `COPY main.py`** — Docker layer caching: if you only change `main.py`, pip install doesn't re-run.

- **`--gpus all`** — exposes all GPUs. For a single-GPU box, fine. For multi-GPU, use `--gpus '"device=0"'`.

## Two non-obvious design choices in the compose

1. **Shared HF cache volume.** Both services mount the same `./hf_cache`. Hugging Face's cache keys by repo, so there's no collision — but if both pulled, e.g., a tokenizer from the same repo, they'd share it. More importantly: one persistent cache dir on the host is easier to back up and reason about than two.

2. **Same env-var name (`SHARED_TOKEN`) feeds different container env-vars (`TTS_TOKEN`, `STT_TOKEN`).** Each service reads its own variable, but you only set one secret in `.env`. Simpler operations, same security (one secret = one thing to rotate). The two services can't be talked to by *each other* — only by the browser extension that holds the token.

## When separate GPUs help vs. when it's overkill

**Combined STT+TTS VRAM is ~5 GB** with the picks we discussed:
- IndicParler-TTS ~1.5 GB
- whisper-large-v3 ~3 GB
- CUDA workspace + headroom ~0.5 GB

So if you have one 16 GB GPU, both services fit comfortably on it with ~10 GB to spare. Putting them on separate GPUs is **technically unnecessary** for VRAM.

**But there are three real benefits to separation, even when one GPU would fit both:**

| Benefit | Why it matters |
|---|---|
| **Failure isolation** | TTS hang / OOM / CUDA error → STT keeps serving. One GPU = one failure domain. |
| **Independent restarts** | You'll iterate on TTS more than STT (voices, streaming tweaks). Restarting a container that shares the GPU briefly stalls the other service's CUDA context. |
| **Concurrency without contention** | If you ever have two users at once, STT inference and TTS inference don't fight for SM cores or memory bandwidth. Latency is steadier. |

**The downsides:**

| Downside | Mitigation |
|---|---|
| Wastes idle GPU capacity | If you have 2 GPUs available anyway, no real waste |
| Asymmetric load — STT busy, TTS idle | Only matters at scale; not v1 |

**My take:** if you have 2+ GPUs available, separate them. Cleaner ops for free. If you have only one big GPU, share it — one fits five.

## How to do it (Docker mechanics)

Two ways to pin a container to a specific GPU:

**Option 1: `device_ids` in compose (clean, idiomatic):**

```yaml
services:
  tts:
    # ...
    deploy:
      resources:
        reservations:
          devices:
            - driver: nvidia
              device_ids: ["0"]      # ← GPU 0
              capabilities: [gpu]

  stt:
    # ...
    deploy:
      resources:
        reservations:
          devices:
            - driver: nvidia
              device_ids: ["1"]      # ← GPU 1
              capabilities: [gpu]
```

**Option 2: `CUDA_VISIBLE_DEVICES` env var (works everywhere):**

```yaml
services:
  tts:
    environment:
      CUDA_VISIBLE_DEVICES: "0"
    # still need --gpus all or count: all to expose GPUs
  stt:
    environment:
      CUDA_VISIBLE_DEVICES: "1"
```

**One mental-model gotcha:** inside each container, the assigned GPU appears as `cuda:0` regardless of which physical GPU it is. So your Python code always uses `cuda:0` — Docker handles the mapping. Don't hardcode `cuda:1` inside the container; you'll get a "device not found" error.

**Verification once running:**

```bash
docker exec tts-container nvidia-smi   # should show only GPU 0
docker exec stt-container nvidia-smi   # should show only GPU 1
```

If each container sees exactly one GPU and they're different ones — assignment worked.

## A subtler trade-off: GPU heterogeneity

If your two GPUs are *different* (e.g., one T4 + one A10), put the **bigger model on the bigger GPU**. With our picks:

- **whisper-large-v3 (STT)** → bigger GPU. Larger model, batch-mode inference, latency-sensitive.
- **IndicParler-TTS** → smaller GPU. Smaller model, sentence-by-sentence work.

If they're the same model GPU, it doesn't matter which gets which.

3. Transcribe both wav files. For Hindi, pass `language="hi"` explicitly the first time to remove auto-detect from the equation.

**Success criteria:**
- English wav → English transcription, close to the input.
- Hindi wav → Hindi transcription in Devanagari.
- Inference time: < 1 second for ~5 sec audio.

**Common gotchas:**
- *cuDNN / cuBLAS errors*: faster-whisper uses CTranslate2, which has stricter version requirements than PyTorch. If you see `libcudnn.so.9: cannot open shared object file`, your host cuDNN is older than CT2 expects. Two fixes: (a) `pip install nvidia-cudnn-cu12==9.*` to get cuDNN into the venv, or (b) just move to A.4 (container) where the PyTorch image bundles the right versions.
- *`device_index=0`* — this is where you control which GPU. For host testing, GPU 0 is fine. In the container, you'll let Docker assign and use `device_index=0` always (since the container only sees one GPU).

Good question — and it points at a really common confusion worth clearing up cleanly. Let me separate three layers that are easy to conflate.

## The three layers

```
┌────────────────────────────────────────────────────────────┐
│ Layer 3: HTTP API surface                                  │
│   "POST /v1/audio/speech returns audio"                    │
│   ── this is your main.py + FastAPI ──                     │
└────────────────────────────────────────────────────────────┘
              ▲ uses
┌────────────────────────────────────────────────────────────┐
│ Layer 2: Model inference                                   │
│   "weights + forward pass + tokenizer"                     │
│   ── parler_tts / faster_whisper Python libraries ──       │
└────────────────────────────────────────────────────────────┘
              ▲ runs on
┌────────────────────────────────────────────────────────────┐
│ Layer 1: GPU + container                                   │
│   "CUDA driver, VRAM, kernel scheduling"                   │
│   ── Docker + nvidia-container-toolkit ──                  │
└────────────────────────────────────────────────────────────┘
```

The **GPU does not expose an API**. Neither does the **model**. A GPU is just compute hardware; a model is a bag of floats + a forward function. You need a **program** (your `main.py`) to translate "HTTP POST → call model → serialize result → HTTP response."

So yes — **each service needs its own `main.py`**, because STT and TTS:
- Load different models
- Expose different endpoints (`/v1/audio/transcriptions` vs `/v1/audio/speech`)
- Have different request/response shapes (multipart audio in vs JSON text in)

What Docker *does* give you "for free":
- GPU isolation (`device_ids: ["0"]` makes only that GPU visible)
- Port forwarding (container's 8001 → host's 8001)
- Process lifecycle (restart on crash)

What it doesn't give you:
- Anything about HTTP routes, request validation, streaming responses, auth — that's all `main.py`.

## "Wait — don't some model servers auto-expose an API?"

Yes, and your question is reasonable because this exists for some model families:

| Auto-API server | What it serves | Why we're not using it |
|---|---|---|
| **vLLM** | LLMs (OpenAI-compatible) | LLMs only — no Whisper or Parler-TTS |
| **TGI** (HF Text Generation Inference) | LLMs | Same |
| **Triton Inference Server** | Anything with TensorRT/ONNX | Setup is heavy; we'd have to convert models |
| **faster-whisper-server** (community) | Whisper only | Could use for STT — but not for TTS, so you'd still write half the stack |
| **LitServe** (Lightning) | Anything, wraps your model | Adds a framework dependency for ~80 lines of FastAPI |

For **just LLM serving**, you'd indeed use vLLM and get a free OpenAI-compatible endpoint. Your `memexpaas-llm-api` running Qwen-30B is almost certainly vLLM doing exactly that. But for STT + TTS, the ecosystem is patchier, and writing a small FastAPI app is genuinely the simplest path — about 80 lines per service, full control, no surprise behaviors.

## `tts/main.py`

```python
import os
import re
import time
import asyncio
import logging
from contextlib import asynccontextmanager
from typing import Optional

import torch
import numpy as np
import lameenc
from fastapi import FastAPI, Header, HTTPException
from fastapi.responses import StreamingResponse
from pydantic import BaseModel

from parler_tts import ParlerTTSForConditionalGeneration
from transformers import AutoTokenizer

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
log = logging.getLogger("tts")

MODEL_ID = "ai4bharat/indic-parler-tts"
DEVICE = "cuda:0"   # container sees only its assigned GPU as cuda:0
TOKEN = os.environ.get("TTS_TOKEN", "")

# Named voices → free-text descriptions IndicParler understands.
# Extend this catalog over time; clients just pass the name.
VOICE_DESCRIPTIONS = {
    # English
    "alloy":   "A clear female voice with a calm, neutral tone speaking at a natural pace, very clean recording.",
    "echo":    "A warm male voice with a confident tone, speaking moderately fast and clearly.",
    "nova":    "A bright female voice, expressive and energetic, speaking at a natural pace.",
    "onyx":    "A deep male voice, calm and measured, speaking slowly and clearly.",
    "shimmer": "A soft female voice with a gentle tone, speaking at a relaxed pace.",
    # Hindi
    "hf_alpha": "एक स्पष्ट महिला आवाज़, शांत और प्राकृतिक गति से बोलती हुई, बहुत स्पष्ट रिकॉर्डिंग।",
    "hm_alpha": "एक गहरी पुरुष आवाज़, आत्मविश्वास के साथ मध्यम गति से बोलती हुई।",
}
DEFAULT_DESCRIPTION = VOICE_DESCRIPTIONS["alloy"]

# Model handles (filled in lifespan)
model = None
tokenizer = None
description_tokenizer = None
SAMPLE_RATE = 24000
inference_lock = asyncio.Lock()


@asynccontextmanager
async def lifespan(app: FastAPI):
    global model, tokenizer, description_tokenizer, SAMPLE_RATE
    log.info(f"loading {MODEL_ID} on {DEVICE}...")
    t0 = time.time()
    model = ParlerTTSForConditionalGeneration.from_pretrained(
        MODEL_ID, torch_dtype=torch.float16
    ).to(DEVICE)
    tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
    description_tokenizer = AutoTokenizer.from_pretrained(
        model.config.text_encoder._name_or_path
    )
    SAMPLE_RATE = model.config.sampling_rate
    log.info(f"loaded in {time.time()-t0:.1f}s; sample_rate={SAMPLE_RATE}")
    yield
    log.info("shutting down")


app = FastAPI(lifespan=lifespan)


def check_auth(authorization: Optional[str]):
    if not TOKEN:
        log.warning("TTS_TOKEN not set; rejecting all requests")
        raise HTTPException(503, "server not configured")
    if not authorization or not authorization.startswith("Bearer "):
        raise HTTPException(401, "missing bearer token")
    if authorization.split(" ", 1)[1] != TOKEN:
        raise HTTPException(401, "invalid token")


@app.get("/health")
async def health():
    vram_mb = 0
    if torch.cuda.is_available():
        vram_mb = torch.cuda.memory_allocated() // (1024 * 1024)
    return {
        "model_loaded": model is not None,
        "model": MODEL_ID,
        "sample_rate": SAMPLE_RATE,
        "vram_mb": vram_mb,
        "voices": list(VOICE_DESCRIPTIONS.keys()),
    }


class SpeechRequest(BaseModel):
    model: str = "local-tts"
    input: str
    voice: str = "alloy"
    response_format: str = "mp3"


def split_sentences(text: str):
    # Devanagari danda (।) included for Hindi
    parts = re.split(r"(?<=[.!?।])\s+", text.strip())
    return [p for p in parts if p]


def synth_sentence(prompt: str, description: str) -> np.ndarray:
    """Blocking inference for one sentence. Returns float32 audio."""
    desc_ids = description_tokenizer(description, return_tensors="pt").to(DEVICE)
    prompt_ids = tokenizer(prompt, return_tensors="pt").to(DEVICE)
    with torch.inference_mode():
        gen = model.generate(
            input_ids=desc_ids.input_ids,
            attention_mask=desc_ids.attention_mask,
            prompt_input_ids=prompt_ids.input_ids,
            prompt_attention_mask=prompt_ids.attention_mask,
        )
    return gen.cpu().numpy().squeeze().astype(np.float32)


async def stream_mp3(text: str, voice: str):
    """Yield MP3 bytes, sentence-by-sentence, through one persistent encoder."""
    description = VOICE_DESCRIPTIONS.get(voice, DEFAULT_DESCRIPTION)

    encoder = lameenc.Encoder()
    encoder.set_bit_rate(128)
    encoder.set_in_sample_rate(SAMPLE_RATE)
    encoder.set_channels(1)
    encoder.set_quality(2)  # 0=highest, 9=lowest

    sentences = split_sentences(text) or [text]
    loop = asyncio.get_running_loop()
    t_start = time.time()
    t_first = None

    for i, sent in enumerate(sentences):
        # Push blocking inference to a thread so the event loop stays alive
        audio = await loop.run_in_executor(None, synth_sentence, sent, description)

        # float32 [-1, 1] → int16 PCM
        pcm = np.clip(audio, -1.0, 1.0)
        pcm = (pcm * 32767).astype(np.int16).tobytes()
        mp3_chunk = encoder.encode(pcm)
        if mp3_chunk:
            if t_first is None:
                t_first = time.time() - t_start
                log.info(f"TTFB {t_first*1000:.0f}ms (sentence 1/{len(sentences)})")
            yield bytes(mp3_chunk)

    tail = encoder.flush()
    if tail:
        yield bytes(tail)
    log.info(f"synth done: {len(sentences)} sentences in {time.time()-t_start:.2f}s")


@app.post("/v1/audio/speech")
async def speech(req: SpeechRequest, authorization: Optional[str] = Header(None)):
    check_auth(authorization)
    if model is None:
        raise HTTPException(503, "model not loaded")
    if req.response_format != "mp3":
        raise HTTPException(400, f"only mp3 supported in v1, got {req.response_format}")
    if not req.input.strip():
        raise HTTPException(400, "empty input")

    async def guarded_stream():
        async with inference_lock:
            async for chunk in stream_mp3(req.input, req.voice):
                yield chunk

    return StreamingResponse(guarded_stream(), media_type="audio/mpeg")
```

## `stt/main.py`

```python
import os
import io
import time
import asyncio
import logging
from contextlib import asynccontextmanager
from typing import Optional

import torch
from fastapi import FastAPI, Header, HTTPException, UploadFile, File, Form
from fastapi.responses import PlainTextResponse

from faster_whisper import WhisperModel

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
log = logging.getLogger("stt")

MODEL_SIZE = "large-v3"
COMPUTE_TYPE = "float16"
DEVICE = "cuda"
DEVICE_INDEX = 0  # always 0 inside container — docker maps the physical GPU
TOKEN = os.environ.get("STT_TOKEN", "")

model: Optional[WhisperModel] = None
inference_lock = asyncio.Lock()


@asynccontextmanager
async def lifespan(app: FastAPI):
    global model
    log.info(f"loading whisper-{MODEL_SIZE} on {DEVICE}:{DEVICE_INDEX} ({COMPUTE_TYPE})...")
    t0 = time.time()
    model = WhisperModel(
        MODEL_SIZE,
        device=DEVICE,
        device_index=DEVICE_INDEX,
        compute_type=COMPUTE_TYPE,
    )
    log.info(f"loaded in {time.time()-t0:.1f}s")
    yield


app = FastAPI(lifespan=lifespan)


def check_auth(authorization: Optional[str]):
    if not TOKEN:
        log.warning("STT_TOKEN not set; rejecting all requests")
        raise HTTPException(503, "server not configured")
    if not authorization or not authorization.startswith("Bearer "):
        raise HTTPException(401, "missing bearer token")
    if authorization.split(" ", 1)[1] != TOKEN:
        raise HTTPException(401, "invalid token")


@app.get("/health")
async def health():
    vram_mb = 0
    if torch.cuda.is_available():
        vram_mb = torch.cuda.memory_allocated() // (1024 * 1024)
    return {
        "model_loaded": model is not None,
        "model": MODEL_SIZE,
        "vram_mb": vram_mb,
    }


@app.post("/v1/audio/transcriptions")
async def transcribe(
    file: UploadFile = File(...),
    model_name: Optional[str] = Form("whisper-1", alias="model"),
    language: Optional[str] = Form(None),
    response_format: str = Form("json"),
    prompt: Optional[str] = Form(None),
    authorization: Optional[str] = Header(None),
):
    check_auth(authorization)
    if model is None:
        raise HTTPException(503, "model not loaded")

    audio_bytes = await file.read()
    if not audio_bytes:
        raise HTTPException(400, "empty audio")

    audio_io = io.BytesIO(audio_bytes)
    t0 = time.time()

    async with inference_lock:
        def run():
            segments, info = model.transcribe(
                audio_io,
                language=language,        # None = auto-detect
                vad_filter=True,          # trims silence
                initial_prompt=prompt,
                beam_size=5,
            )
            segs = list(segments)         # consume generator
            text = "".join(s.text for s in segs).strip()
            return text, info
        text, info = await asyncio.get_running_loop().run_in_executor(None, run)

    duration = time.time() - t0
    log.info(
        f"transcribed: lang={info.language} (p={info.language_probability:.2f}), "
        f"audio_dur={info.duration:.1f}s, infer={duration:.2f}s, chars={len(text)}"
    )

    if response_format == "text":
        return PlainTextResponse(text)
    return {
        "text": text,
        "language": info.language,
        "duration": info.duration,
    }
```

## A few things to know before you deploy

1. **First-run downloads are big.** IndicParler weights are ~2 GB, whisper-large-v3 ~3 GB. Both go into `hf_cache/` which is bind-mounted; subsequent restarts are instant.
2. **`parler-tts` from git** — if pip refuses (e.g., behind a proxy that blocks git protocol), clone the repo separately and `COPY` it in, then `pip install /repo`.
3. **TTFB on the very first request is slow** (5–15 s) due to CUDA graph warmup. The *second* request shows real numbers. You can warm with a dummy synth at the end of `lifespan` if startup latency matters.
4. **`lameenc` doesn't always emit on the first `encode()` call** if PCM is shorter than a frame (~50 ms). For sentences >0.5 s this is never an issue.
5. **`vad_filter=True` in STT** depends on Silero VAD weights bundled with faster-whisper. Should work out of the box; if it 503s on startup, drop the flag.
6. **No rate limiting / no retry logic.** Add if you expose this to anyone besides yourself.
7. **TTS request body size** — IndicParler tops out around ~600 chars per sentence. If your LLM produces enormously long sentences, split them at commas too.

When you've got the curl tests working end-to-end (especially: round-tripping TTS audio through STT and getting recognizable text), the GPU half of the project is done.

Good — let's add this properly. There's a real choice between Cloudflare Tunnel and Caddy, and they have meaningfully different operational properties. I'll recommend one, then show both.

## The choice

| | **Cloudflare Tunnel** | **Caddy + Let's Encrypt** |
|---|---|---|
| Public IP needed? | **No** — outbound tunnel | Yes — A records → your box |
| Ports to open on your box? | **None** | 80 + 443 |
| Behind NAT / corporate firewall? | Works fine | Won't work |
| TLS certs | Cloudflare manages | Caddy auto-renews from LE |
| DDoS protection | Free | None |
| You control the path entirely? | No — traffic flows via CF edge | Yes |
| Extra auth (SSO/OTP) | Available via Cloudflare Access (free) | DIY |
| Latency penalty | ~10–30 ms (CF edge proximity) | None |
| Failure mode if Cloudflare has an outage | You're down | Unaffected |

**My recommendation: Cloudflare Tunnel.** Three reasons:

1. **Zero firewall config** — your GPU box stays fully closed to the internet. The tunnel daemon makes an outbound connection. This is genuinely the safest posture for a home/office-hosted GPU server.
2. **You already own the domain on Cloudflare** (implied by `rahulsaraf.online`), so DNS, tunnel, and certs all live in one place.
3. **Adding Cloudflare Access later is one toggle** — you can require Google SSO before anyone even hits the bearer-token check. Belt-and-suspenders without extra code.

The downside (latency + dependency on CF) is real but small for a voice service where you're already in the 300–500 ms range.

```yaml
services:
  stt:
    build: ./stt
    expose: ["8002"]                  # internal only — no host port mapping
    environment:
      STT_TOKEN: ${SHARED_TOKEN}
    volumes:
      - ./hf_cache:/cache/hf
    deploy:
      resources:
        reservations:
          devices:
            - driver: nvidia
              device_ids: ["0"]
              capabilities: [gpu]
    restart: unless-stopped

  tts:
    build: ./tts
    expose: ["8001"]                  # internal only
    environment:
      TTS_TOKEN: ${SHARED_TOKEN}
    volumes:
      - ./hf_cache:/cache/hf
    deploy:
      resources:
        reservations:
          devices:
            - driver: nvidia
              device_ids: ["1"]
              capabilities: [gpu]
    restart: unless-stopped

  cloudflared:
    image: cloudflare/cloudflared:latest
    command: tunnel --no-autoupdate run
    environment:
      TUNNEL_TOKEN: ${TUNNEL_TOKEN}
    depends_on:
      - stt
      - tts
    restart: unless-stopped
```

**Three subtle changes worth noticing:**

1. **`ports:` → `expose:`** on `stt` and `tts`. Previously we mapped to host ports `8002:8002` and `8001:8001`; now they're container-internal. **Only cloudflared talks to them**, via Docker's default bridge network. The host firewall doesn't need any inbound rules.
2. **`depends_on`** — cloudflared waits until stt/tts containers are *started* (not necessarily ready). That's fine; HTTP requests routed before model load will get 503s, and Cloudflare will retry. For stricter sequencing you'd add healthchecks.
3. **No `volumes:` or extra config files needed for cloudflared.** The token contains everything — tunnel ID, account ID, route rules are fetched at startup. Much simpler than the old `config.yml` approach.

Notice: no port number, automatic HTTPS, no certbot, no DNS waiting. That's the appeal.

## Caddy alternative (only if you don't want Cloudflare)

If you specifically need to avoid Cloudflare (data sovereignty, latency, etc.), here's the equivalent with Caddy. Requires:

- A public IP for your GPU box
- Ports 80 + 443 forwarded to it
- A records for `stt.rahulsaraf.online` and `tts.rahulsaraf.online` pointing at that IP

### `caddy/Caddyfile`

```
stt.rahulsaraf.online {
    reverse_proxy stt:8002
}

tts.rahulsaraf.online {
    reverse_proxy tts:8001
}
```

### Caddy service in compose

```yaml
  caddy:
    image: caddy:latest
    ports:
      - "80:80"
      - "443:443"
    volumes:
      - ./caddy/Caddyfile:/etc/caddy/Caddyfile:ro
      - caddy_data:/data
      - caddy_config:/config
    depends_on:
      - stt
      - tts
    restart: unless-stopped

volumes:
  caddy_data:
  caddy_config:
```

Caddy auto-fetches Let's Encrypt certs on first hit and renews them. The `caddy_data` volume persists the certs across restarts (important — LE rate-limits aggressively).

You'd switch `expose:` back to keeping the services internal, just like with cloudflared.

## One security upgrade you should consider (Cloudflare path only)

Right now, anyone on the internet who finds `tts.rahulsaraf.online` can hit the endpoint — they just can't get past the bearer token. That's fine, but cloudflared lets you add **Cloudflare Access** for ~5 minutes of work and zero code:

1. Zero Trust → Access → Applications → Add an application → Self-hosted
2. Application domain: `*.rahulsaraf.online`
3. Add a policy: "Allow if email is `rahul1.saraf@ril.com`" (or your Google SSO group)

After this, hitting the URL in a browser prompts for SSO login before *anything* — including the bearer-token check — runs. The bearer token still protects programmatic access (the extension passes it via header; CF Access detects the service-token header and bypasses SSO).

This is genuinely worth setting up for any service you expose. But it's optional for v1, especially if the bearer token stays out of git.

*Cost: $0.1191 | Duration: 28.7s*

Excellent catch — and you're right to be confused, because the compose file *doesn't* assign URLs. The mapping lives somewhere else.

## Where the URL routing actually lives

With the **token-based tunnel** approach we chose, the routing config is stored **inside Cloudflare's dashboard**, not in your codebase. The `TUNNEL_TOKEN` is essentially a pointer that says "I am tunnel `voice-stack` in account X" — and Cloudflare looks up the routing rules for that tunnel.

So the flow is:

```
1. You create the tunnel in dashboard          → CF assigns it an ID + token
2. You add hostnames in dashboard              → CF stores routing rules
3. cloudflared starts with TUNNEL_TOKEN        → it fetches rules from CF
4. Request to tts.rahulsaraf.online            → CF edge routes to your tunnel
5. cloudflared receives the request            → matches hostname against rules
6. Forwards to http://tts:8001                 → docker DNS resolves "tts"
```

The two non-obvious pieces:

| Question | Answer |
|---|---|
| How does Cloudflare know `tts.rahulsaraf.online` should go to your tunnel? | DNS — when you added the public hostname, CF created a CNAME for `tts` pointing at `<tunnel-id>.cfargotunnel.com`. |
| How does cloudflared know which service to forward to? | It fetches the ingress rules from Cloudflare on startup (and on changes). The rules say "host=tts.rahulsaraf.online → service=http://tts:8001". |
| How does cloudflared reach the `tts` container by name? | Docker compose puts all services on a shared bridge network with built-in DNS. `tts` and `stt` resolve as hostnames inside that network. |

So the answer to your question: **the URL → service mapping is configured by you in the Cloudflare Zero Trust dashboard, not in docker-compose.yml.** The compose file just spins up the agent that connects to that pre-configured tunnel.

## If you'd rather have routing in your repo

There's an alternative pattern — **config-file based tunnel** — where the routing rules live in a local YAML file you check in. Some teams prefer this because the config is grep-able, diff-able, and reviewable.

It looks like this:

```yaml
# cloudflared/config.yml
tunnel: <tunnel-uuid>
credentials-file: /etc/cloudflared/creds.json

ingress:
  - hostname: stt.rahulsaraf.online
    service: http://stt:8002
  - hostname: tts.rahulsaraf.online
    service: http://tts:8001
  - service: http_status:404      # required catch-all
```

And the compose service:

```yaml
  cloudflared:
    image: cloudflare/cloudflared:latest
    command: tunnel --config /etc/cloudflared/config.yml run
    volumes:
      - ./cloudflared/config.yml:/etc/cloudflared/config.yml:ro
      - ./cloudflared/creds.json:/etc/cloudflared/creds.json:ro
    depends_on: [stt, tts]
    restart: unless-stopped
```

The `creds.json` file is the secret part — you'd generate it once via `cloudflared tunnel login` + `cloudflared tunnel create voice-stack` on your laptop, then copy it to the GPU box. It's the same trust unit as the token, just stored as a file. Should never be checked into git.

DNS still has to be set up — either in the dashboard, or via `cloudflared tunnel route dns voice-stack tts.rahulsaraf.online`.

## Comparison

| | **Token (current)** | **Config file** |
|---|---|---|
| Routing config lives in | Cloudflare dashboard | Your repo (`config.yml`) |
| Adding a new subdomain | Click in dashboard | Edit YAML, restart cloudflared |
| Visible in `git log` / code review | No | Yes |
| Secret management | One env var | One file mount |
| Mutate from anywhere (mobile, etc.) | Yes | No, need ssh access |
| Best when | Solo / small team, comfortable in dashboard | Multi-engineer, want config in code |

## My honest recommendation for your case

**Stick with the token approach.** Three reasons:

1. **One-engineer setup right now.** Routing config in a dashboard is fine when only one person changes it.
2. **You can always migrate later.** The tunnel ID stays the same — you just stop using the token and start using credentials + config file. No data loss, no downtime if done carefully.
3. **Fewer secret files to manage.** The token in `.env` is one secret to rotate; the config-file path has both the creds.json *and* knowledge of the tunnel UUID.

If this ever grows to a team where multiple people change routing, switch to config-file. Until then, the dashboard is the simpler tool.

Short answer: **we're not — not in this file.** The URLs are assigned in the Cloudflare dashboard, and the token is just a pointer that says "use *this* tunnel's pre-configured rules."

## What the token actually contains

`TUNNEL_TOKEN` is a base64-encoded JSON blob with three things:

```
{ "a": "<account-id>",
  "t": "<tunnel-uuid>",
  "s": "<tunnel-secret>" }
```

That's it. It does **not** contain "stt.rahulsaraf.online → stt:8002". It just lets cloudflared authenticate as a specific tunnel.

## Where the URL → service mapping lives

When you created the tunnel and added public hostnames in the dashboard, Cloudflare stored a record like this **server-side**, attached to the tunnel UUID:

```yaml
# (this lives on Cloudflare's side, not in your repo)
ingress:
  - hostname: stt.rahulsaraf.online
    service: http://stt:8002
  - hostname: tts.rahulsaraf.online
    service: http://tts:8001
  - service: http_status:404
```

And Cloudflare also added two DNS records:

```
stt.rahulsaraf.online   CNAME   <tunnel-uuid>.cfargotunnel.com
tts.rahulsaraf.online   CNAME   <tunnel-uuid>.cfargotunnel.com
```

## The request lifecycle

```
1. Browser → DNS lookup for tts.rahulsaraf.online
              ↓ (resolves to Cloudflare edge via the CNAME)
2. Browser → HTTPS to Cloudflare edge
              ↓
3. CF edge → "this hostname belongs to tunnel <uuid>"
              ↓ (over the persistent outbound connection cloudflared
              ↓  established when it started)
4. cloudflared (in your container) receives the request
              ↓ (looks up ingress rules it fetched from CF on startup)
5. cloudflared → http://tts:8001 (Docker DNS resolves "tts")
              ↓
6. tts container → response
              ↓ (back up the same chain)
7. Browser receives audio
```

The compose snippet you quoted is responsible for **step 4 only**: spin up cloudflared, give it the token, let it phone home. The mapping in steps 3 and 5 was set when you configured the tunnel in the dashboard — and it's stored on Cloudflare's side, retrieved at cloudflared startup.

## TL;DR

| Thing | Where it's defined |
|---|---|
| Tunnel exists | Cloudflare dashboard (one-time, you do it) |
| DNS: `tts.rahulsaraf.online` → tunnel | Cloudflare dashboard (auto-created when you add hostname) |
| Routing: hostname → `http://tts:8001` | Cloudflare dashboard (you fill this in) |
| Auth: which tunnel are we | `TUNNEL_TOKEN` in `.env` |
| Service `tts` resolves to a container | Docker compose's bridge network DNS |

If you'd rather have the routing rules visible in your repo instead of in the dashboard, switch to the config-file pattern (`config.yml` + `creds.json` mounted into the container). Same end result, just different storage location for the rules.

This error has nothing to do with your compose file or our code — it's a host-level NVIDIA driver / library version mismatch. Very common, very fixable.

## What "driver/library version mismatch" means

Two things must agree:

- **Kernel module** (`nvidia.ko`) — loaded into the running kernel
- **User-space libraries** (`libnvidia-ml.so`, etc.) — what nvidia-container-cli reads

After a driver package update, apt installs the new user-space libraries, but the kernel still has the **old** module loaded. They disagree. Result: nvml refuses to initialize. Affects everything — `nvidia-smi`, Docker, your own Python.

## Confirm it's the same problem on the host

Run `nvidia-smi` directly on the host (not in Docker):

```bash
nvidia-smi
```

You'll almost certainly see the same error: `Failed to initialize NVML: Driver/library version mismatch`. That confirms it.

If you want to see the version skew explicitly:

```bash
cat /proc/driver/nvidia/version        # kernel module version
dpkg -l | grep nvidia-driver           # installed user-space version
```

## Fix — three options, easiest first

### Option 1: Reboot (10 seconds of work, definitely fixes it)

```bash
sudo reboot
```

After reboot, `nvidia-smi` works on the host, then `docker compose up -d` succeeds. **This is what I'd do.** Unless you have stateful work running on the box, the reboot is genuinely the cheapest fix and avoids the chance of leaving things in a half-fixed state.

### Option 2: Reload kernel modules without reboot

Only works if **no process is using the GPU** (no X server, no other CUDA apps). Stop docker first, then:

```bash
sudo systemctl stop docker
sudo rmmod nvidia_uvm nvidia_drm nvidia_modeset nvidia
sudo modprobe nvidia
sudo modprobe nvidia_uvm
sudo systemctl start docker
nvidia-smi    # should now work
```

**Watch out:** if `rmmod nvidia` fails with "Module nvidia is in use", something still holds it — find with `lsof /dev/nvidia*` or `fuser -v /dev/nvidia*`. Kill those processes, then retry. If you can't unload, you must reboot.

### Option 3: Fully reinstall drivers (only if reboot doesn't fix it)

Rare, but happens if the package update was interrupted:

```bash
sudo apt update
sudo apt install --reinstall nvidia-driver-535  # or whichever version
sudo reboot
```

## After the host is fixed

Three verification steps before retrying compose:

```bash
# 1. Host-level GPU access
nvidia-smi
# expect: table showing both T10s

# 2. Docker can see GPUs
docker run --rm --gpus all nvidia/cuda:12.1.1-base-ubuntu22.04 nvidia-smi
# expect: same table from inside a container

# 3. Now bring up the stack
cd /app/data/voice-stack
docker compose up -d
docker compose logs -f
```

## What's happening

```
VLLM::Eng (PID 1456950, running as root) ── holds /dev/nvidia1
```

That's almost certainly your **own LLM server** — the vLLM process serving Qwen3-35B at `memexpaas-llm-api.rahulsaraf.online`. It's running on the same physical box as where you want to deploy STT/TTS, and it's been alive since **before** the driver-library skew happened. It's still working because it grabbed GPU handles when the driver was consistent — but the kernel-vs-userspace mismatch means **any *new* process trying to acquire GPU access (including Docker containers) will fail** until things are realigned.

So you're caught between two facts:

1. **You can't fix the driver mismatch** without either rebooting or `rmmod nvidia` — both of which kill vLLM.
2. **You can't bring up new GPU-using containers** while the mismatch exists.

There's no clever workaround. The fix is fundamentally "realign kernel module + libraries," which means vLLM goes down briefly.

## Two important things to check first

Before you do anything, get the full GPU picture:

```bash
# Everything holding any NVIDIA device
sudo lsof /dev/nvidia* 2>/dev/null | awk '{print $1, $2, $NF}' | sort -u

# Or more concise: who's on each GPU
sudo fuser -v /dev/nvidia0 /dev/nvidia1 2>&1
```

You want to know: **is GPU 0 free, or is something on it too?** Your lsof output only showed GPU 1 references for vLLM — if GPU 0 is genuinely idle, that opens an option (see below).

Also worth confirming the version skew:

```bash
cat /proc/driver/nvidia/version    # kernel module
nvidia-smi --version 2>&1 || true  # userspace (will error, but may show version)
```

## Three realistic options

### Option A: Coordinated restart of vLLM + reboot (cleanest)

**Plan a 5-minute window.** Steps:

```bash
# 1. Note how vLLM is launched so you can restart it (systemd unit? screen? docker?)
ps -p 1456950 -o args=

# 2. Stop vLLM gracefully (use whatever launched it — systemctl, kill, etc.)
sudo systemctl stop vllm  # or however it was started

# 3. Reboot
sudo reboot

# 4. After reboot: nvidia-smi works → start vLLM → start voice-stack
```

Once both are up, you have your full 2-GPU layout: vLLM on GPU 1 (or wherever it was), voice-stack on the other GPU. **This is what I'd do.**

### Option B: Module reload without reboot

Slightly less disruptive but more brittle:

```bash
sudo systemctl stop docker
sudo systemctl stop vllm                 # or kill PID 1456950
sudo rmmod nvidia_uvm nvidia_drm nvidia_modeset nvidia
sudo modprobe nvidia
sudo modprobe nvidia_uvm
sudo systemctl start docker
# verify:
nvidia-smi
# restart vLLM and voice-stack
```

Same downtime envelope (a couple of minutes), no full reboot. The risk: if any other process is touching `/dev/nvidia*` and you missed it, `rmmod` fails and you're stuck partway. The `lsof` check above catches that.

If GPU 0 turns out to be completely idle (no one's using it), you can change the compose to put **both STT and TTS on GPU 0** instead of splitting across both GPUs:

## Why STT gets the token too

I added `HF_TOKEN` to both services even though only TTS hit the gate. Reason: `openai/whisper-large-v3` isn't currently gated, but if HF gates it later (it has happened with other popular models) you'd be in the same situation. Passing the token now costs nothing and avoids a future re-debug. `huggingface_hub` ignores the env var when the model isn't gated.

## How `huggingface_hub` picks it up

You don't have to change `main.py`. The library checks `HF_TOKEN` (and the older `HUGGING_FACE_HUB_TOKEN`) env vars automatically when making API calls. Passing the env var into the container is the entire integration.

Great — the GPU half is up. Now to wire the URLs. This is all in the Cloudflare dashboard; no more code changes.

## Prerequisite: verify the tunnel is actually connected

Before adding hostnames, confirm cloudflared registered itself. On the GPU box:

```bash
docker compose logs cloudflared | grep -i "registered\|error" | tail -10
```

You want to see four lines like:

```
INF Registered tunnel connection connIndex=0 ... location=BOM/MAA/...
INF Registered tunnel connection connIndex=1 ...
INF Registered tunnel connection connIndex=2 ...
INF Registered tunnel connection connIndex=3 ...
```

If you see errors instead (especially "no such tunnel" or auth issues), your `TUNNEL_TOKEN` is wrong — go back and copy it again. If you see the four `Registered` lines, you're good.

## Step 1 — Find your tunnel in the dashboard

1. Go to https://one.dash.cloudflare.com (the Zero Trust dashboard).
2. Make sure the **account selector** in the top-left is on the same account that owns `rahulsaraf.online`. If you have multiple accounts, this matters — the tunnel lives in one specific account.
3. Left sidebar → **Networks** → **Tunnels**.
4. You should see your tunnel (named `voice-stack` or whatever you called it). Its status should say **HEALTHY** with a green dot. If it says **INACTIVE**, the container isn't talking to Cloudflare — check the `docker compose logs cloudflared` output again.
5. Click on the tunnel name to open it.

## Step 2 — Add the STT public hostname

You'll be on the tunnel's detail page. There are four tabs at the top — **Overview, Public Hostname, Private Network, Configure**.

1. Click the **Public Hostname** tab.
2. Click **Add a public hostname**.
3. Fill out the form:

| Field | Value | Why |
|---|---|---|
| **Subdomain** | `stt` | The `stt.` part of `stt.rahulsaraf.online` |
| **Domain** | `rahulsaraf.online` | Dropdown — your domain |
| **Path** | (leave blank) | Routes everything under the subdomain |
| **Type** | `HTTP` | Cloudflare talks plaintext to the container; TLS is terminated at CF edge |
| **URL** | `stt:8002` | `stt` = the docker-compose service name; `8002` = the port inside the container |

**Three crucial things to get right** (these are where people trip up):

- The **Type is `HTTP`, not `HTTPS`**. Your containers don't have TLS certificates — cloudflared connects plaintext over the docker network. Cloudflare still serves HTTPS to the public; the encryption ends at the edge.
- The **URL is `stt:8002`, not `localhost:8002`**. Each container has its own network namespace; `localhost` inside cloudflared = cloudflared itself. Docker compose's built-in DNS resolves `stt` and `tts` as hostnames on the shared network.
- **No `http://` prefix in the URL field** — Cloudflare's UI adds it based on the Type dropdown. If you type `http://stt:8002` it'll become `http://http://stt:8002` and fail.

4. Click **Save hostname**.

**What just happened behind the scenes:**
- Cloudflare created a CNAME DNS record: `stt.rahulsaraf.online` → `<tunnel-uuid>.cfargotunnel.com`
- Cloudflare added an ingress rule to the tunnel's config: incoming requests for that hostname route to `http://stt:8002`
- Cloudflared (in your container) sees the config change within seconds and starts accepting traffic for that hostname

## Step 3 — Add the TTS public hostname

Same process, different values:

| Field | Value |
|---|---|
| **Subdomain** | `tts` |
| **Domain** | `rahulsaraf.online` |
| **Path** | (blank) |
| **Type** | `HTTP` |
| **URL** | `tts:8001` |

Click **Save hostname**.

You should now see two entries on the Public Hostname tab.

## Step 4 — Verify DNS propagation

From any machine (your laptop, the GPU box, anywhere):

```bash
dig stt.rahulsaraf.online +short
dig tts.rahulsaraf.online +short
```

You should see output like:

```
abc123def456.cfargotunnel.com.
104.21.X.X
172.67.X.X
```

If you see `NXDOMAIN` or no answer, wait 30 seconds and try again. Cloudflare DNS is usually instant but occasionally takes a minute.

| Symptom | Likely cause | Fix |
|---|---|---|
| `curl: (6) Could not resolve host` | DNS hasn't propagated yet | Wait 30s, retry |

Worth setting up before you give the URL to anyone else, but not necessary for personal use today.

Two things are showing in this log, and they tell an interesting story:

## What's actually happening

1. **`WARN[0000] The "TUNNEL_TOKEN" variable is not set.`** — docker-compose can't interpolate `${TUNNEL_TOKEN}` from your `.env`. Either the token is missing/placeholder, the `.env` file isn't being read, or the variable name is wrong.

2. **The tunnel that IS running has stale ingress rules pointing at `caddy:80`** — a service that doesn't exist on your voice-stack network. The errors show real internet traffic hitting `rahulsaraf.online` and getting routed to a `caddy` service that isn't there. **This is leftover routing from a previous project.**

In other words: your cloudflared *did* connect to a Cloudflare tunnel (the container is up and accepting traffic from CF edge), but it connected to **a pre-existing tunnel from an older setup**, and that tunnel's ingress rules in the dashboard still point at a Caddy-fronted thing from before.

## Diagnostic — let's get clarity

Run these and share the output:

```bash
cd /.blob/rahuketu/programming/voice-stack

# 1. What's in .env? Is TUNNEL_TOKEN actually set to a real value?
cat .env

# 2. Is cloudflared running, and how long has it been up?
docker compose ps

# 3. How was cloudflared started — what env did it actually see?
docker inspect $(docker compose ps -q cloudflared) | grep -A 5 '"Env"'
```

The `inspect` step is the key one. The container's stored env shows what it actually started with — that tells us whether `TUNNEL_TOKEN` was set when the container was created (it would still be running on that even if `.env` has changed since).

## What I think you'll find

One of two scenarios:

**Scenario A: `.env` has the placeholder**, never got the real token, but cloudflared was started with a token via some other means (maybe earlier you ran `docker compose up` with a different env, or set `TUNNEL_TOKEN` in your shell).

**Scenario B: `.env` has the real token**, and that token belongs to a tunnel you previously created for a different project (the one with caddy). cloudflared is dutifully connecting to *that* tunnel, picking up its old ingress rules, and trying to reach `caddy`.

Either way, the fix is the same: align the tunnel's dashboard config with what voice-stack actually serves.

How to refresh cloudflare token

There's no "rotate this token" button for dashboard-managed tunnels in Cloudflare's UI — the token effectively *is* the tunnel's credentials. To get a fresh token, the cleanest path is to **delete the tunnel and create a new one**. It takes about 2 minutes.

## Why deletion is OK

You're not losing anything important — a Cloudflare Tunnel object stores:
- An ID (changes on recreation)
- Public hostnames + their routing rules (you re-add — 30 seconds)
- The DNS CNAMEs (auto-recreate when you re-add the hostnames)
- The token (the thing you want to refresh)

Nothing else. No traffic history, no logs are lost (those live separately in Cloudflare's analytics).

## Steps

### 1. Delete the old tunnel

Cloudflare dashboard → **Networks → Tunnels** → click the three-dots menu next to your tunnel → **Delete tunnel**.

Confirm. The tunnel's DNS records and ingress rules are deleted with it. The CNAMEs for `stt.rahulsaraf.online` and `tts.rahulsaraf.online` (if you'd added them) go away too.

### 2. Create a fresh tunnel

Still in **Networks → Tunnels** → **Create a tunnel**.

- Connector type: **Cloudflared**
- Tunnel name: `voice-stack` (or whatever)
- On the **Install connector** screen, you'll see the new token. **Copy it now** — Cloudflare won't show it again later.

### 3. Add the public hostnames (in the same flow)

The wizard's next screen lets you add hostnames immediately:

| Subdomain | Domain | Service |
|---|---|---|
| `stt` | `rahulsaraf.online` | `HTTP` `stt:8002` |
| `tts` | `rahulsaraf.online` | `HTTP` `tts:8001` |

Save. DNS auto-created.

### 4. Update `.env` with the new token

```bash
cd /.blob/rahuketu/programming/voice-stack
nano .env       # or vim/your editor
```

Replace the old `TUNNEL_TOKEN=...` value with the new one you just copied.

### 5. Restart cloudflared with the new token

```bash
docker compose up -d --force-recreate cloudflared
docker compose logs cloudflared | tail -20
```

You want to see the four `Registered tunnel connection` lines, no `caddy` errors.

### 6. Verify

```bash
curl https://tts.rahulsaraf.online/health
curl https://stt.rahulsaraf.online/health
```

Both should return JSON.

Recreating the tunnel kills two birds:
- Fresh token (your stated goal)

## What each flag is doing

| Flag | Why it's there |
|---|---|
| `--build` | Rebuild images if `Dockerfile`s or source changed. Cheap if nothing changed (Docker's layer cache makes it a no-op). |
| `--force-recreate` | Replace running containers even if compose thinks they're up-to-date. Necessary when env vars change — without this, the old container keeps running with the old `TUNNEL_TOKEN` baked in at create time. |
| `-d` | Detached. |

## Three things to know

1. **Rotate this token now.** You pasted it into this transcript, which means it's in the chat log. For a learning/dev box that's probably fine, but for hygiene's sake regenerate via the Cloudflare dashboard (delete + create new tunnel, as we discussed), and put the *new* one in `.env` instead.

2. **You don't actually need `--build` unless code/Dockerfile changed.** Plain env-var swap → `docker compose up -d --force-recreate cloudflared` (just that one service, faster) is enough. Use `--build` only after editing a `Dockerfile` or `main.py`.

3. **Inline env vars leak to shell history** (`~/.bash_history` or `~/.zsh_history`). The `.env` approach avoids that. To scrub the history afterward: `history -d <N>` for the specific entry, or just `unset HISTFILE && exec bash` to start a fresh history-less shell.

| `127.0.0.1:8002` | `stt:8002` |

| `stt:8002/` (trailing slash) | `stt:8002` |

## If you want to use a different filename (not `.env`)

Two ways to point compose at a non-default env file:

**Flag (per-invocation):**
```bash
docker compose --env-file path/to/myenv.env up -d
```

**Permanent (in compose.yml) — pass file contents into a specific container:**
```yaml
services:
  cloudflared:
    env_file:
      - ./secrets/cloudflared.env
```

But `env_file:` only makes vars available **inside that container** — it does NOT help compose-level interpolation like `${TUNNEL_TOKEN}`. For interpolation you must use either the default `.env` or `--env-file`.

Now we have a clear answer. The log shows two phases:

1. QUIC failed (UDP blocked, as expected) → cloudflared **auto-fell-back to HTTP/2** without needing my `--protocol http2` flag. Good.
2. HTTP/2 connection got: **`Unauthorized: Invalid tunnel secret`** — repeated forever.

So your `TUNNEL_TOKEN` is invalid. The tunnel ID `dfe53a71-f7fa-4e44-8cb1-2d3bafb8ee78` in the log matches what you posted earlier, so the tunnel exists — but the **secret part** of the token has gone stale. Most likely cause: you deleted and recreated the tunnel in the dashboard (during the "refresh token" step), and the running container is still running with the **first** token, not the **new** one.

## Verify the mismatch

```bash
cd /.blob/rahuketu/programming/voice-stack

# What's in .env?
grep TUNNEL_TOKEN .env | head -c 100; echo

# What did the running container actually start with?
docker compose exec cloudflared printenv TUNNEL_TOKEN 2>/dev/null | head -c 100; echo
```

If these two values differ, that's the entire problem — the container is running with a stale token. If they're the same, the token itself is wrong.

## Fix

### Step 1 — Get the current valid token

The most reliable path: **delete and recreate the tunnel once more**, copying the token *immediately* after creation, before doing anything else.

1. Dashboard → **Networks → Tunnels** → find your tunnel → three-dots menu → **Delete tunnel** (confirm).
2. **Create a tunnel** → Connector type: Cloudflared → name: `voice-stack`.
3. On the **Install connector** screen, you'll see the install command with the token. **Copy the entire base64 string** (between `--token ` and the end). Don't truncate.
4. Add the two public hostnames in the same wizard:
   - `stt.rahulsaraf.online` → HTTP → `stt:8002`
   - `tts.rahulsaraf.online` → HTTP → `tts:8001`

### Step 2 — Update `.env` with the fresh token

```bash
cd /.blob/rahuketu/programming/voice-stack

# Edit interactively
nano .env

# OR set it via sed (replace YOUR_NEW_TOKEN with the actual one):
sed -i 's|^TUNNEL_TOKEN=.*|TUNNEL_TOKEN=YOUR_NEW_TOKEN|' .env

# Verify it's there
grep TUNNEL_TOKEN .env
```

### Step 3 — Force-recreate cloudflared so it picks up the new token

```bash
docker compose up -d --force-recreate cloudflared
docker compose logs -f cloudflared
```

This time you should see:

```
INF Starting tunnel tunnelID=<new-uuid>
INF Registered tunnel connection connIndex=0 ...
INF Registered tunnel connection connIndex=1 ...
INF Registered tunnel connection connIndex=2 ...
INF Registered tunnel connection connIndex=3 ...
```

No `Invalid tunnel secret`. No QUIC errors (it'll likely auto-fall-back to HTTP/2 again, which is fine).

### Step 4 — Verify end-to-end

```bash
curl https://stt.rahulsaraf.online/health
curl https://tts.rahulsaraf.online/health
```

Both should return JSON now.

1. **QUIC fallback worked automatically** — the `Switching to fallback protocol http2` line shows cloudflared handled UDP-being-blocked on its own. My earlier `--protocol http2` flag was preventive, not required. (You can keep the flag; it just shortens the connection time by skipping the QUIC attempt.)

Yes — **same machine is fine, no special assignment needed**. Cloudflare tunnels aren't bound to a machine; they're just an identity (UUID + secret). A "connector" is whatever cloudflared process authenticates with that identity, from wherever.

So your setup will be:

| Machine | Container | Connects to | Reaches backends |
|---|---|---|---|
| GPU box | old cloudflared (caddy stack's compose) | OLD tunnel | caddy network |
| GPU box | voice-stack's cloudflared | NEW tunnel | voice-stack_default |

Two cloudflared containers running side-by-side on the same host, each connecting to a different tunnel. Cloudflare has zero concept of "machine assignment" — it just sees two independent connectors.

## What happens automatically

When you create the new tunnel and add `stt` + `tts` as public hostnames, Cloudflare:

1. Generates new tunnel UUID + secret → embedded in the token you copy
2. Auto-creates CNAME records: `stt.rahulsaraf.online → <new-uuid>.cfargotunnel.com` and same for `tts`
3. Routes any incoming request for those hostnames to whichever connector is authenticated against this **specific** tunnel ID

Your existing CNAMEs for `webtop`, `jupyter`, etc. still point to `<old-uuid>.cfargotunnel.com` and continue routing through the old tunnel. They don't move.

## What you have to do manually

The only thing you must do on the machine is:

1. Update `voice-stack/.env` with the new token
2. `docker compose up -d --force-recreate cloudflared`

No firewall changes, no DNS work (Cloudflare manages DNS for `rahulsaraf.online`), no port forwarding. The voice-stack cloudflared makes an outbound connection to Cloudflare's edge — same as the old one — and announces "I'm a connector for tunnel `<new-uuid>`."

## One thing to verify after

Make sure the OLD cloudflared isn't still claiming `stt` and `tts`. After Step 2 in the plan (deleting `stt`/`tts` from the old tunnel's Public Hostnames), the old tunnel's ingress should no longer list them. If both tunnels claim the same hostname, Cloudflare's behavior is "last writer wins" at the DNS level and weird intermittent routing — easy to avoid by just keeping the lists clean.

Go ahead with the plan as written — it'll work on the same box without any extra steps.

## A useful frame in the meantime

Whatever the answers, the redesign comes down to **two surfaces** + **one transport choice**:

| Surface | What changes |
|---|---|
| **Outgoing audio capture** | Browser `MediaRecorder` API → blob → POST as `multipart/form-data` to `/v1/audio/transcriptions`. The OpenAI client SDK does this for you, but a raw `fetch` is also fine. |
| **Incoming audio playback** | POST JSON to `/v1/audio/speech` → response body is an MP3 stream → feed into an `<audio>` element or `AudioBuffer`. |

The **transport choice** is whether you stream or batch:
- **Batch** (what `test.sh` does): record full utterance → POST → wait → play. Simple, ~1–3s latency. Good first version.
- **Streaming**: chunk audio as it's captured (VAD-segmented), POST chunks, play partial replies. ~500ms latency, way more code, requires changes to the server too (your current TTS endpoint isn't streaming-friendly out of the box).

Start with batch. Optimize later only if the UX demands it.

The second is what Gradio naturally gives you, and it's also the cleanest way to share code between "Chrome extension" and "Gradio app". The extension becomes a thin shell that talks to your backend; the backend talks to STT, solveit, and TTS.

Gradio's `gr.Audio(sources=["microphone"], type="filepath")` already gives you push-to-talk recording with a file at the end. So a Gradio app is basically:

## The mental model first

A Chrome extension isn't one thing. It's **3–4 different JavaScript runtimes** that talk to each other via message passing, each with different powers:

| Runtime | What it can do | What it can't |
|---|---|---|
| **Popup** (popup.html) | Full DOM, chrome.* APIs, persistent UI when open | Dies when popup closes |
| **Content script** | Reads/writes the host page's DOM | No access to the page's JS variables (isolated world); limited chrome.* |
| **Page context script** (injected `<script>`) | Full access to host page's JS (e.g. `window.htmx`) | No chrome.* APIs |
| **Service worker** (background.js) | Long-running, chrome.* full access, network, alarms | No DOM, can be killed and restarted |
| **Options page** (options.html) | Like popup but full-tab | — |

Most "real" extensions use 2–3 of these. The voice extension you read uses popup + content script + injected page script. The existing extension is a good *reference* — let's not copy it but understand why it's split the way it is.

# Milestone 1 — "Hello, extension"

## What you'll build

A toolbar-icon popup with a single button. Each click increments a counter that persists across browser restarts. About 30 lines of code total.

## What you'll learn

| Concept | Why it matters |
|---|---|
| **manifest.json** | The "contract" — every extension must have this. Declares name, version, permissions, entry points. |
| **manifest v3** | Current spec. v2 is deprecated. Different in subtle ways (service workers instead of background pages, etc.). |
| **chrome.action** | The toolbar button. The popup is its `default_popup`. |
| **chrome.storage.local** | The right way to persist data. `localStorage` works inside popups but it's scoped to the popup origin — `chrome.storage` is the proper API and async. |
| **The dev loop** | Edit files → reload at `chrome://extensions` → re-open popup. You'll be doing this hundreds of times. |
| **Why popup state evaporates** | The popup is destroyed when closed. Anything you stored in JS variables is gone. Storage API is how you survive. |

## File layout

Three files. That's it.

```
memexsolve-voice/
├── manifest.json     # the "contract"
├── popup.html        # what the popup looks like
└── popup.js          # what the popup does
```

## The dev loop (do this first, before any code)

1. Open Chrome → navigate to `chrome://extensions`
2. Top-right toggle: enable **Developer mode**
3. New buttons appear top-left. Click **Load unpacked**
4. Select the folder `memexsolve-voice` (currently empty — that's fine, you'll see an error, that's the point of doing this first)
5. The error tells you Chrome can't find `manifest.json`. Good — now we'll create it.

After every edit, you'll come back here and click the **🔄 reload** icon on the extension card. **The popup won't pick up changes until you reload here.**

## Step 1 — Write `manifest.json`

I'll give you this one verbatim because it's mostly config, not logic. Type it out (don't copy-paste) — getting the field names into muscle memory matters.

```json
{
  "manifest_version": 3,
  "name": "MemexSolve Voice",
  "version": "0.1.0",
  "description": "Voice modality for SolveIt and MemexSolve.",

  "action": {
    "default_popup": "popup.html",
    "default_title": "MemexSolve Voice"
  },

  "permissions": ["storage"]
}
```

**Read line-by-line:**

- `"manifest_version": 3` — must be 3. v2 is dead.
- `"name"`, `"version"`, `"description"` — shown in `chrome://extensions` and the Chrome Web Store. `version` must be `dot.dot.dot` format.
- `"action"` — declares the toolbar icon. `default_popup` is the HTML file Chrome opens when the icon is clicked.
- `"permissions": ["storage"]` — opt-in to the `chrome.storage` API. Without this, `chrome.storage.local` is undefined.

Notice there are **no icons declared** — Chrome will show a default puzzle-piece icon. We'll add icons later; they're cosmetic.

## Step 2 — Write `popup.html`

Also verbatim, since it's HTML scaffolding. Keep it tiny.

```html
<!DOCTYPE html>
<html>
<head>
  <meta charset="utf-8">
  <style>
    body { width: 200px; padding: 12px; font-family: system-ui; }
    button { width: 100%; padding: 8px; font-size: 14px; cursor: pointer; }
    #count { font-size: 18px; margin-bottom: 8px; text-align: center; }
  </style>
</head>
<body>
  <div id="count">Loading...</div>
  <button id="inc">+1</button>

  <script src="popup.js"></script>
</body>
</html>
```

**Two important rules** that bite everyone:

- ❌ **No inline JS allowed** in extension HTML. `<button onclick="...">` will be **silently ignored** because manifest v3 enforces a strict Content Security Policy. You must wire handlers in `popup.js` using `addEventListener`.
- The `<script>` tag must be loaded *after* the elements it references, OR you must wait for `DOMContentLoaded`. The placement at the bottom of `<body>` is the simplest fix.

## Step 3 — `popup.js` — your turn

Now you write this. The popup should:

1. On load, read the current count from storage and display it in `#count`.
2. When `#inc` is clicked, increment the count, save it, and update the display.

**The two API calls you need:**

```js
// Read a value (async, callback OR promise — both work in MV3)
const result = await chrome.storage.local.get('count');
// → result is { count: <whatever was stored> } or {} if never set

// Write a value
await chrome.storage.local.set({ count: 42 });
```

`chrome.storage.local.get('count')` returns `{}` when the key was never set, not `{ count: undefined }`. So:

Destructuring with a default. If `count` is missing in the result, it's `0`.

Most beginners write this with the read+display logic duplicated (once on load, once after click). Cleaner:

Okay before I start extension, I just remembered one more thing. Should I deploy VAD model as well on server. Old speech pipeline which needs to be modified with new api's looks as below
```python
"""
speech_pipeline.py
VAD (Silero) + ASR (faster-whisper medium) pipeline.

Input:  raw PCM bytes, 16 kHz, 16-bit signed, mono
Output: (transcript string, detected language code e.g. "en", "hi")

Set WHISPER_LANGUAGE env var to force a language (e.g. "en", "hi").
Leave unset for auto-detection — handles English, Hindi, and Hinglish.
"""

import logging
import os
import numpy as np
import torch
from silero_vad import load_silero_vad, get_speech_timestamps
from faster_whisper import WhisperModel

# None = auto-detect; set to "en" or "hi" to force
WHISPER_LANGUAGE: str | None = os.getenv("WHISPER_LANGUAGE") or None

logger = logging.getLogger(__name__)

# ── Model config (from .env) ───────────────────────────────────────────────────
# WHISPER_MODEL: "medium" (default, fits M1 16 GB), "large-v3" (GPU, 3× accuracy)
WHISPER_MODEL: str = os.getenv("WHISPER_MODEL", "medium")

# ── Device selection ───────────────────────────────────────────────────────────
# faster-whisper uses CTranslate2 which supports cuda and cpu.
# MPS is not supported by CTranslate2 — CPU is used on Apple Silicon.
# int8 quantisation keeps memory low (~500 MB) and speeds up CPU inference.
_CT2_DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
_CT2_COMPUTE = "float16" if torch.cuda.is_available() else "int8"
logger.info(f"Whisper device: {_CT2_DEVICE}, compute: {_CT2_COMPUTE}, model: {WHISPER_MODEL}")

# ── Silero VAD ─────────────────────────────────────────────────────────────────
_vad_model = None

def _load_vad():
    global _vad_model
    if _vad_model is None:
        logger.info("Loading Silero VAD...")
        _vad_model = load_silero_vad()
        _vad_model.eval()
        logger.info("Silero VAD loaded.")


def strip_silence(pcm_bytes: bytes, sample_rate: int = 16000) -> np.ndarray:
    """Remove silent frames using Silero VAD. Returns float32 speech-only audio."""
    _load_vad()

    audio_np = np.frombuffer(pcm_bytes, dtype=np.int16).astype(np.float32) / 32768.0
    audio_tensor = torch.from_numpy(audio_np)

    speech_timestamps = get_speech_timestamps(
        audio_tensor,
        _vad_model,
        sampling_rate=sample_rate,
        threshold=0.5,
        min_speech_duration_ms=200,
        min_silence_duration_ms=300,
    )

    if not speech_timestamps:
        logger.debug("VAD: no speech detected")
        return np.array([], dtype=np.float32)

    segments = [audio_tensor[ts["start"]: ts["end"]].numpy() for ts in speech_timestamps]
    speech_audio = np.concatenate(segments)
    logger.debug(f"VAD: {len(speech_timestamps)} segments, {len(speech_audio)/sample_rate:.1f}s speech")
    return speech_audio


# ── Whisper ASR ────────────────────────────────────────────────────────────────
_asr_model: WhisperModel | None = None

def _load_asr():
    global _asr_model
    if _asr_model is None:
        logger.info(
            "Loading Whisper model",
            extra={"model": WHISPER_MODEL, "device": _CT2_DEVICE},
        )
        _asr_model = WhisperModel(
            WHISPER_MODEL,
            device=_CT2_DEVICE,
            compute_type=_CT2_COMPUTE,
        )
        logger.info("Whisper ASR loaded.")


def transcribe(pcm_bytes: bytes, sample_rate: int = 16000) -> tuple[str, str]:
    """PCM bytes → VAD strip silence → Whisper → (transcript, language_code)."""
    _load_asr()

    speech_audio = strip_silence(pcm_bytes, sample_rate)
    if len(speech_audio) < sample_rate * 0.3:
        logger.info("ASR: audio too short after VAD, skipping")
        return "", "en"

    segments, info = _asr_model.transcribe(
        speech_audio,
        language=WHISPER_LANGUAGE,  # None = auto-detect
        beam_size=5,
        vad_filter=False,  # Silero VAD already handled this
    )
    transcript = " ".join(seg.text.strip() for seg in segments).strip()
    logger.info(
        f"ASR transcript: '{transcript}' "
        f"(lang={info.language}, prob={info.language_probability:.2f})"
    )
    return transcript, info.language
```

Great question to pause on before building. Let me lay out the trade-off space, because the answer depends on what UX you want and most engineers get this wrong on the first try.

## Where can VAD live? Three options.

| Where | Used for | Latency |
|---|---|---|
| **Server, before Whisper** | Strip silence to make Whisper run faster | Adds 50–200ms per request |
| **Inside Whisper itself** (faster-whisper's `vad_filter=True`) | Same as above, but bundled | 0ms extra — same model loads VAD once |
| **In the browser/client** | Real-time UX: detect speech start/end, auto-stop recording, draw mic activity meter | 10–20ms per chunk, local |

Notice these aren't alternatives — they solve *different* problems.

## Map to your use cases

### For push-to-talk (your stated UX)

The user holds a button → speaks → releases. So:

- "Strip silence" is **not critical** — Whisper transcribes whether or not silence is present. Difference is maybe 200ms of inference time.
- You already know when speech starts (button down) and ends (button up). Client-side VAD adds nothing.
- A separate VAD service round-trip would *increase* latency, not reduce it.

➜ **For push-to-talk, you don't need a dedicated VAD server.** Either (a) enable `vad_filter=True` in the existing STT server (free — faster-whisper bundles Silero internally), or (b) skip VAD entirely. The accuracy difference is negligible.

### For "auto-stop when I stop talking" (smart push-to-talk, hands-free dictation)

Here client-side VAD becomes valuable: you draw a mic-activity meter, auto-stop after 1.5s of silence, etc.

But you'd run VAD **in the browser**, not on the server. Server VAD would mean streaming chunks over the network at 100ms intervals just to ask "is this speech?" — terrible round-trip economics. The browser has compute to spare; use it.

The library you'd use: [`@ricky0123/vad-web`](https://github.com/ricky0123/vad). It runs **Silero VAD as ONNX in WebAssembly** right in the browser. Same model you have on the server, but offline-capable, zero network cost.

### For wake-word / always-on listening

Same answer: browser-side. You don't want to stream audio to your server 24/7 just to detect when the user says "hey memex".

**Now (Phase 1):**
- Don't deploy a VAD service.
- Check your existing STT server — it's probably already using faster-whisper. If `vad_filter=True` is set on the `transcribe()` call, you're done. If not, flip it on.

Good moment to pause and review before duplicating the architecture. I've read all of it. Let me split this into three parts: **review of what's there**, **a critical correction about Voxtral**, and **the design decisions you need to make before adding stt2/tts2**.

## Part 1 — Review of the existing stack

This is genuinely well-built. Calling out the patterns worth keeping (so you carry them into stt2/tts2) and a few nits.

### Things that are right ✅

| Pattern | Where | Why it matters |
|---|---|---|
| **OpenAI-compatible surface** | both servers | drop-in replacement for `openai-python`, no client-side adapter needed |
| **Single bearer token + 401 on mismatch** | `check_auth` | minimal but correct; same token across services via `SHARED_TOKEN` |
| **Lifespan loads model at startup** | both `lifespan()` | first request is fast, not 5–30s slow |
| **`asyncio.Lock` around inference** | both | serializes GPU calls — no OOM races on concurrent requests |
| **`run_in_executor` for blocking work** | both `_run` / `_synth_pcm` | the event loop stays responsive for `/health` and queuing |
| **Sentence-by-sentence TTS streaming** | `_stream_mp3` | TTFB ~500ms instead of 5s — huge UX win |
| **Single persistent MP3 encoder** | `encoder = _make_encoder()` outside loop | no inter-sentence pops/clicks (an easy mistake to make) |
| **GPU isolation via compose** | `device_ids: ["0"]`, `["1"]` | STT and TTS can't starve each other |
| **HF cache as a bind mount** | `./hf_cache:/cache/hf` | rebuilding the container doesn't re-download 3 GB |
| **Warmup call at startup** | TTS `lifespan` | dodges the 10s cold-start on the first real request |

### Things I'd polish (none urgent)

1. **STT has no `temperature` clamping**: line 136 does `temperature or 0.0` — that's fine, but a client could pass e.g. `2.0` and get hallucinations. Probably fine for trusted use.
2. **No `/v1/models` endpoint**: some OpenAI clients (especially LangChain) call this on init. Easy to add: return `{"data": [{"id":"whisper-1","object":"model"}]}`.
3. **Token comparison `!=`**: not constant-time. Timing-attack-vulnerable in theory; effectively a non-issue for self-hosted behind Cloudflare. Skip unless you want to be paranoid.
4. **`inference_lock` is module-level**: works because uvicorn creates one event loop per process, but a more idiomatic place is inside `lifespan()` so it's bound to the loop. Cosmetic.
5. **TTS hard-codes `cuda:0`**: line 35. The container only sees its assigned card as `cuda:0`, so this is correct — but if you ever run two TTS replicas in one container, this would need to change.

### One subtle thing worth understanding

In TTS, `_stream_mp3` calls `loop.run_in_executor(None, _synth_pcm, sent, description)` **inside the for-loop**, sentence by sentence. This means:

- Sentence N's GPU work blocks sentence N+1's GPU work (within one request) — which is fine, you only have one GPU.
- But it also means the encoder is **CPU-side and sequential** while the GPU is idle between sentences. You could overlap them (pipeline: GPU synthesizes sentence N+1 while CPU encodes N), but it's not worth the complexity here.

That kind of micro-pipelining is what production TTS services do (e.g. Cartesia, ElevenLabs). For your scale, the simple loop is the right call.

I have to flag this before you go further:

## Part 3 — Design decisions before any code

Before scaffolding `stt2/` and `tts2/` directories, four things to settle:

### Decision A — GPU allocation

Your existing stack uses GPU 0 (STT) and GPU 1 (TTS). Adding stt2 + tts2 means:

| Option | Setup | Trade-off |
|---|---|---|
| **One GPU per service (4 total)** | Need 4 cards; ideal isolation | Requires hardware |
| **Share GPU per role**: stt+stt2 on GPU 0, tts+tts2 on GPU 1 | Both ASR services share a card | Risk of OOM if both run inference simultaneously |
| **Pair by quiet pattern**: stt on GPU 0, tts on GPU 1, stt2 on GPU 1, tts2 on GPU 0 | Each card runs one ASR + one TTS | Cross-stack contention |
| **Reuse GPU 0 for stt2 only** (skip tts2 for now) | Voxtral-Mini is 3B, ~7 GB VRAM in fp16; fits beside whisper-large-v3 (~3 GB) on a 16 GB card | Pragmatic |

How many GPUs does your box have and what are their VRAM sizes? `nvidia-smi -L` will tell us.

### Decision B — Same OpenAI API surface, or different?

**Voxtral's killer feature** is audio-grounded Q&A, not just transcription:

```
POST /v1/audio/transcriptions   ← whisper-style: audio → text
```

vs Voxtral can also do:

```
POST /v1/audio/chat             ← audio + text prompt → text answer
{ "audio": <upload>, "messages": [{"role":"user","content":"summarize this"}] }
```

You could either:
- **Wrap Voxtral behind the same `/v1/audio/transcriptions` endpoint** → simple drop-in alt for stt
- **Expose a richer endpoint** that uses Voxtral's full multimodal LLM capability → unlocks new UX in your extension/Gradio app

The second is more interesting *if* your downstream apps will use it. Otherwise it's wasted potential.

### Decision C — Hostname / routing

Two hostnames per service, like:

```
stt.rahulsaraf.online   → faster-whisper
stt2.rahulsaraf.online  → Voxtral
tts.rahulsaraf.online   → IndicParler
tts2.rahulsaraf.online  → (whichever)
```

Add them to the voice-stack tunnel's Public Hostnames (the dedicated tunnel we set up earlier). Each new service = new compose service + new ingress entry + new container port.

### Decision D — Runtime: vLLM vs. transformers

For Voxtral specifically, Mistral recommends `vllm[audio]`. It's a different beast than transformers:

| | transformers (what your TTS uses) | vLLM (Mistral's recommendation for Voxtral) |
|---|---|---|
| Loading | `from_pretrained(...)` | `vllm.LLM(...)` |
| Server | You write FastAPI yourself | vLLM ships its own OpenAI-compatible server |
| Performance | Good for single requests | Much better under concurrent load (paged attention) |
| Surface | Custom code | `python -m vllm.entrypoints.openai.api_server ...` |

If you use vLLM, `stt2/main.py` may become essentially **just a thin Dockerfile wrapping vLLM's existing server** — barely any custom Python. That's a different kind of project than the hand-rolled stt/tts you have today.

Smart question — and the answer is "yes-ish, but probably not in the way you're hoping." Let me lay out the technical reality on T10 specifically, because this is a good chance to understand how quantization actually works.

## What T10 supports (and doesn't)

| Precision | T10 hardware? | Notes |
|---|---|---|
| FP32, FP16 | ✅ | Standard tensor cores |
| BF16 | ❌ | Ampere+ only |
| **FP8** | **❌** | **Hopper (H100) / Ada Lovelace (RTX 40) only** |
| INT8 | ✅ | Turing has INT8 tensor cores |
| **INT4** | ✅ (sort of) | Turing introduced INT4 tensor cores, but in practice... |

**FP8 is off the table** — your hardware physically can't do it. So one of your two options is dead immediately.

## How "INT4" actually works in practice

This is the part most people get wrong. Modern INT4 quantization (AWQ, GPTQ) **does NOT do INT4 matmul on the GPU.** It does this:

```
Weights stored in VRAM as INT4   →   on-the-fly dequant to FP16   →   FP16 matmul   →   FP16 result
```

So the win is **memory footprint**, not compute speed. You fit a 4B model in ~2 GB of weights instead of 8 GB. But the actual matrix multiplication runs in FP16 — same speed as if you'd loaded the model in FP16.

(There are true INT4-compute kernels — Marlin, ExLlamaV2 — but they require sm_80+ for the fastest paths. On Turing they fall back to dequant-then-FP16.)

## The VRAM math if it worked

| Component | FP16 size | INT4 weights size |
|---|---|---|
| 3.4B transformer LM | 6.8 GB | 1.7 GB |
| 390M flow-matching acoustic | 0.8 GB | (keep FP16: 0.8 GB) |
| 300M neural codec | 0.6 GB | (keep FP16: 0.6 GB) |
| KV cache (streaming) | 2–4 GB | 2–4 GB |
| Activations / workspace | 1–2 GB | 1–2 GB |
| **Total** | **~12–14 GB** | **~6–9 GB** ✅ |

Yes, INT4-quantized Voxtral TTS would fit comfortably on a 16 GB T10. Notice I only quantize the **LM transformer** — the flow-matching and codec stay FP16. More on why in a moment.

## The five reasons this is still not the answer

### 1. Pre-quantized weights don't exist

This is the killer. It's March 2026 — Voxtral TTS shipped weeks ago. Search HuggingFace for `Voxtral-4B-TTS-AWQ` or `-GPTQ` — there's nothing yet. To use INT4 you'd have to **quantize it yourself**, which means:

- Set up AWQ or GPTQ pipeline
- Build a calibration dataset (for TTS this means representative *text* samples spanning your target languages)
- Run quantization (hours to days on T10)
- A/B test output quality vs the FP16 original
- Tune scale/zero-point per layer if quality drops

You'd be the first person doing this. No reference numbers, no Stack Overflow answers, you're the pioneer.

### 2. TTS quantization is genuinely harder than LLM quantization

Quantizing an LLM, errors look like wrong words occasionally. Quantizing a TTS, errors look like:

- Buzzy / metallic timbre
- Robotic prosody
- Clicks and glitches at sentence boundaries
- Mispronunciations
- Volume modulation artifacts

**Humans hear quantization artifacts in audio much more easily than they read them in text.** This is why most production TTS systems run at FP16 minimum.

### 3. The flow-matching acoustic model is the fragile bit

Voxtral TTS is three models stapled together:

```
text → [3.4B LM]  →  [390M flow-matching]  →  [300M codec]  →  audio
       (semantic)    (acoustic, sensitive)    (waveform)
```

The flow-matching component runs iteratively (similar to a small diffusion model) — each step's error feeds the next. Quantizing it accumulates noise. **You can quantize the LM safely, but the acoustic component really wants FP16.** Which is what the table above does — but it means the savings are only on the 3.4B portion, not the whole 4B.

### 4. vLLM Omni quantization support is uncharted

Regular vLLM supports AWQ/GPTQ/Marlin for text LLMs. vLLM Omni (the audio variant) is **brand new** — March 2026. Whether its audio-output kernels accept quantized LM weights, nobody's published a benchmark. Could just work; could need patches.

### 5. The license doesn't disappear

CC-BY-NC 4.0 still applies to derivative works. A quantized version of Voxtral TTS weights is a derivative work. Same non-commercial restriction.

## Where this leaves you

| Path | Reality |
|---|---|
| FP8 Voxtral TTS on T10 | ❌ Hardware doesn't support FP8 |
| INT4 Voxtral TTS on T10 | ⚠️ Possible as a research project (weeks), not a deployment |
| FP16 Voxtral TTS on T10 | ⚠️ ~12–14 GB, tight; would need BF16→FP16 cast which may go NaN; vLLM Omni Turing support unclear |
| Voxtral TTS hosted by Mistral | ✅ Works today; $0.016/1k chars; no license issue if you pay |
| Kokoro / F5-TTS / XTTS on T10 | ✅ Fits natively, permissive licenses, deploy in an afternoon |

## The mental model to take away

**Quantization is a memory trick, not a magic compatibility trick.** It lets you fit bigger models on smaller cards by sacrificing weight precision. It does *not*:

- Make unsupported precisions (FP8) available
- Make audio quality preserve itself
- Make missing pre-quantized weights appear
- Override licenses

The right time to reach for quantization is **after** you have:
- Pre-quantized weights from a reputable source
- Hardware whose tensor cores match the quantized format
- A use case where the quality drop is acceptable
- A baseline FP16 deployment to A/B against

Voxtral TTS on T10 currently has zero of those four.

Good question — this is the right thing to think about. The answer is **it changes one specific constraint and leaves the others intact.** Let me walk through what an extra pair of T10s unlocks, and what it doesn't.

## What having more T10s actually unlocks: tensor parallelism

The single biggest constraint on a lone T10 isn't compute, it's the **16 GB VRAM ceiling** per card. The way around that ceiling — and you can do this even with your *existing* machine, no need for a second box — is **tensor parallelism (TP)**.

### How TP works (mental model)

Imagine the model is a big stack of matrices. With TP=2:

```
Layer 1's weight matrix W (2048×2048)
  →  Card 0 holds left half  (2048×1024)
  →  Card 1 holds right half (2048×1024)

At inference:
  →  Card 0 computes its half-result
  →  Card 1 computes its half-result
  →  AllReduce sums them back together  ← network/PCIe traffic
  →  Move to next layer
```

You **double your effective VRAM** at the cost of inter-card communication every layer. vLLM has first-class support for this — just `--tensor-parallel-size 2`.

### Voxtral TTS VRAM math with TP=2 on 2× T10

| Item | Single T10 (FP16) | 2× T10 with TP=2 (FP16) |
|---|---|---|
| Weights per card | ~8 GB | ~4 GB |
| KV cache | 2–4 GB | 1–2 GB per card |
| Activations | 1–2 GB | 1 GB per card |
| **Per-card total** | **~12–14 GB** (tight, no headroom) | **~6–7 GB** (lots of headroom) ✅ |

So yes, **on 2× T10 in tensor-parallel mode, Voxtral TTS fits comfortably in FP16.** This is genuinely the situation that changes.

## What still doesn't change

A second machine doesn't solve any of these:

| Constraint | Why TP / extra cards don't help |
|---|---|
| **CC-BY-NC license** | Hardware doesn't change copyright |
| **No BF16 on Turing** | Still have to cast BF16 weights → FP16; quality risk unchanged |
| **No FP8 on Turing** | Hardware-level, can't be parallelized around |
| **T10 is slow** | TP improves *capacity* not *speed*; per-token compute is still T10-class |
| **vLLM Omni's Turing support is unverified** | Same software risk |
| **No NVLink on T10** | TP allreduce goes over PCIe — slower, more overhead than NVLink |

The most important non-VRAM concern is the **NVLink absence**. Let me explain why.

### The NVLink-vs-PCIe issue

Tensor parallelism does an AllReduce **after every transformer layer**. On a 32-layer model with TP=2, that's 32 inter-card communications per forward pass.

| Interconnect | Bandwidth | TP overhead |
|---|---|---|
| NVLink 4 (H100) | ~900 GB/s | ~5–10% latency penalty |
| NVLink 3 (A100) | ~600 GB/s | ~10–15% |
| **PCIe Gen 3 x16 (T10's likely link)** | **~16 GB/s** | **~30–60% latency penalty** |

So on T10 with TP=2, expect **30–60% more latency** than the single-card baseline would predict. The model fits, but it's slower than the math suggests.

Combine that with T10's already-slow tensor cores (vs H200's 70ms reference), and you're looking at **~1–2 seconds latency per sentence**. Acceptable for non-real-time use; not great for snappy push-to-talk.

## Cross-machine TP? No, don't.

If you're hoping all 4 T10s (2 per box) can pool: **don't go there.** Tensor parallelism across machines requires:

- High-bandwidth interconnect (InfiniBand / RoCE / RDMA) — Ethernet works but kills latency
- Same-version software stack, careful network config
- vLLM cross-node TP is mostly used for *training*, not inference
- vLLM Omni cross-node is uncharted territory

You can run **separate vLLM instances** on each machine (no shared model state), which is the same as running two independent services. That's fine for redundancy or A/B testing; it's not "double the VRAM."

## So: does the second machine change my recommendation?

**Yes, mildly.** Specifically:

### Option A — Dedicate the new machine to Voxtral TTS via TP=2

```
Machine 1 (existing):
  GPU 0: stt  (faster-whisper)
  GPU 1: tts  (IndicParler)

Machine 2 (new):
  GPU 0+1: tts2 (Voxtral-4B-TTS via vLLM Omni, TP=2)   ← all the VRAM goes here
```

This unblocks Voxtral TTS technically. You still own the license and quality concerns.

### Option B — Reshuffle to balance both machines

```
Machine 1:
  GPU 0+1: tts2 (Voxtral TTS, TP=2)    ← unblock Voxtral via parallelism

Machine 2:
  GPU 0: stt + stt2 (whisper + Voxtral ASR)
  GPU 1: tts (IndicParler)              ← keep the existing safe TTS
```

This gives you both whisper *and* Voxtral ASR on the same card (they share well), keeps the proven IndicParler as the fallback TTS, and lets Voxtral TTS have the whole TP=2 capacity on machine 1. Cleanest architecture if you want to seriously test Voxtral TTS.

### Option C — Skip Voxtral TTS, use the second machine for something useful

The second machine is genuinely capable. Possibilities:

- **A larger TTS** like F5-TTS Large or XTTS-v2 with extra concurrency capacity
- **Voxtral-Small-24B ASR** instead of Mini — much higher quality ASR via TP=2 (the 24B variant needs ~50 GB FP16, won't fit on 2× T10 but quantized AWQ would fit easily and AWQ versions of Voxtral-Small-24B *do* exist on HuggingFace)
- **A dedicated agent/LLM service** for the voice-stack's brain — Llama-70B AWQ fits on 2× T10
- **Spare capacity** for the rest of memexsolve

Honestly, **Option C is what I'd pick** unless Voxtral TTS specifically is the goal. Two T10s with TP=2 are much more interesting deployed against a 24B ASR or a 70B-quantized chat LLM than against a 4B TTS whose license you can't commercially use.

| Parameters | **82M** | 880M | 4B |
| Disk size (FP16) | ~165 MB | ~1.8 GB | ~8 GB |
| Disk size (INT8) | **~80 MB** | n/a | n/a |
| GPU needed? | **No** | Yes (~6 GB VRAM) | Yes (16+ GB VRAM) |
| License | **Apache 2.0** ✅ | Apache 2.0 ✅ | CC-BY-NC ⚠️ |
| Real-time CPU? | **Yes (~2× faster than real-time)** | No | No |
| Browser deployable? | **Yes** | No | No |

At 82M parameters, Kokoro is small enough that it fits in **the same VRAM budget as a single web page's JavaScript heap**. That's the unlock.

## Deployment surface 1: directly in the browser

This is the wild one. Kokoro has been ported to ONNX and there are first-class JS packages:

| Package | What it gives you |
|---|---|
| `kokoro-js` (npm) | Drop-in wrapper, ~5 lines to synthesize audio |
| `@huggingface/transformers` (Transformers.js) | More flexible, uses ONNX Runtime Web under the hood |
| `onnxruntime-web` | Raw ONNX — most control, most code |

Sketch of what it looks like to use:

```js
import { KokoroTTS } from "kokoro-js";

const tts = await KokoroTTS.from_pretrained("onnx-community/Kokoro-82M-v1.0-ONNX", {
  dtype: "q8",        // int8 quantized → ~80 MB download
  device: "webgpu",   // or "wasm" for CPU fallback
});

const audio = await tts.generate("Hello from the browser", { voice: "af_alloy" });
audio.save("hello.wav");   // or feed to an <audio> element
```

That's it. **No server. No API call. Text → audio, all in the browser.**

### What "webgpu" vs "wasm" actually means

- **WebGPU**: uses the GPU on the user's machine. Chrome / Edge / Safari Tech Preview support it. ~2–5× faster than CPU.
- **WASM (WebAssembly)**: pure CPU fallback. Works in every modern browser. Slower but real-time on any laptop made in the last 5 years.

The library auto-detects and falls back. You don't have to branch.

### What it costs the user

| Cost | Magnitude |
|---|---|
| **First-load download** | 80–300 MB (one-time, cached by browser) |
| **RAM during use** | ~150–400 MB |
| **CPU during synthesis** | ~real-time (1 second of audio takes ~0.5–1s on a modern laptop) |
| **GPU during synthesis** (if WebGPU) | minimal, ~200 MB VRAM, near-instant |

The 80–300 MB initial download is the only sharp edge — show a "first time setup, downloading voice…" UI. After that, it's cached in IndexedDB and instant on subsequent loads.

## Deployment surface 2: CPU on a server

Same story without the browser:

| Package | Use case |
|---|---|
| `kokoro` (PyPI) | Pure Python, PyTorch backend, CPU or CUDA |
| `kokoro-onnx` (PyPI) | ONNX Runtime, CPU-only, faster than PyTorch on CPU |

A few lines of Python:

```python
from kokoro_onnx import Kokoro
kokoro = Kokoro("kokoro-v1.0.onnx", "voices.bin")
samples, sample_rate = kokoro.create(
    "Hello from a CPU-only Docker container",
    voice="af_alloy",
)
```

Speed: **~1–2× real-time on a single modern CPU core.** A 5-second sentence synthesizes in 3–5 seconds. With 4 cores, you can batch.

## What this could change about your architecture

This is where it gets interesting. You've been planning a server-side TTS endpoint for the extension. Kokoro lets you flip the architecture:

### Current plan
```
Extension → MediaRecorder → POST audio → STT server → text
                                                       ↓
                                              solveit/memexsolve
                                                       ↓
Extension ← stream MP3 ← TTS server ← reply text
```

### Browser-Kokoro plan
```
Extension → MediaRecorder → POST audio → STT server → text
                                                       ↓
                                              solveit/memexsolve
                                                       ↓
Extension → in-browser Kokoro → audio (no server, instant)
```

Look what disappears: **no TTS server, no TTS authentication, no streaming protocol, no audio bytes transiting your network, no tts2 deployment at all.** The TTS happens on the user's own machine.

### Pros of browser-Kokoro

| Benefit | Significance |
|---|---|
| Zero TTS server cost / VRAM / ops | Saves a whole service |
| Zero network latency for TTS | Audio plays instantly after text is ready |
| Privacy: text never leaves the user's machine for synthesis | Real win for sensitive work |
| Works offline (after first download) | Genuinely usable on a plane |
| Voice happens at the device, not your cluster | Scales infinitely with no infrastructure |

### Cons / trade-offs

| Trade-off | Mitigation |
|---|---|
| **First-load is 80–300 MB** | Acceptable for an extension; the user installs it once |
| **English-focused** — Hindi voices exist but quality is weaker than IndicParler/Voxtral | Keep IndicParler server-side for Hindi; use Kokoro only for English |
| **Less voice variety than larger models** | Kokoro has ~50 voicepacks; covers most use cases |
| **No voice cloning** | If cloning matters, use F5-TTS server-side |
| **Browser GPU/CPU varies** | WASM fallback ensures it always works, just slower |
| **Older browsers / Chromebooks may struggle** | Falls back to server TTS if Kokoro fails to load |

### A nice hybrid pattern

You don't have to pick one. The extension can:

```
1. Detect Kokoro support on first load.
2. If supported → use browser Kokoro for English.
3. If language is Hindi / unsupported → fall back to server TTS.
4. If Kokoro fails to load → fall back to server TTS.
```

That gives you instant English voice for most users, and the server TTS is only used when needed. Server load drops dramatically.

## When CPU server-side makes sense (vs browser)

| Scenario | Browser-side Kokoro | Server-side Kokoro CPU | Server-side IndicParler GPU |
|---|---|---|---|
| Chrome extension push-to-talk (English) | ✅ Best fit | ⚪ Works | ⚪ Overkill |
| Gradio app on your laptop | ⚪ Possible via Gradio JS hooks | ✅ Simplest | ⚪ Overkill |
| Backend processing of long text (e.g., batch summaries) | ❌ Browser will time out | ✅ Best fit | ⚪ Faster but uses GPU |
| Hindi / multilingual | ❌ Quality drops | ⚪ Marginal | ✅ Best |
| Voice cloning required | ❌ Can't | ❌ Can't | ⚪ Need F5/XTTS |

For your two main targets:
- **Extension push-to-talk** → browser Kokoro is genuinely revolutionary. Try it.
- **Gradio app** → server-side Kokoro on CPU, no GPU service needed for that path.

## Key design decisions encoded in the files

1. **stt2 is config, not code.** The Dockerfile is 15 lines because vLLM ships its own OpenAI-compatible server. All Voxtral specifics (`--tokenizer-mode mistral`, `--gpu-memory-utilization 0.55`, `--enforce-eager` for Turing) live in `docker-compose.yml` where they're visible together.

2. **`--gpu-memory-utilization 0.55` is the critical safety belt.** Without it, vLLM grabs 90% of GPU 0 at startup and OOMs whisper. With 0.55 vLLM caps at ~8.8 GB, leaving whisper its ~3 GB.

3. **`depends_on: stt`** ensures whisper gets its VRAM first. Subtle but matters: vLLM's startup check sees only what's *currently* free, so if it starts first, it'd grab too much.

4. **tts2 has no `deploy.resources`** — that's how you declare "no GPU" in compose. The container runs as a regular CPU process.

5. **Kokoro weights baked into the image** at build time, not downloaded on first request. Adds ~150 MB to the layer but avoids cold-start network calls.

6. **tts3 starts healthy with zero voices.** The catalog is scanned at lifespan startup; if `refs/` is empty, the service runs but rejects synthesis with a helpful 400 message. Lets you deploy first and add voices later.

7. **`vad_filter` and `inference_lock` patterns mirror the existing services** — same async lock around model.generate(), same sentence-level MP3 streaming with a single persistent encoder.

This is **not a voice-stack problem** — it's a host-side NVIDIA driver issue. The good news: your builds all succeeded. The bad news: the host's NVIDIA runtime can't talk to its own kernel module right now.

## What the error actually means

```
nvml error: driver/library version mismatch
```

NVIDIA's kernel module (what's loaded into the running kernel) and NVIDIA's userspace libraries (what `nvidia-container-cli` calls) are at **different versions**. nvml (NVIDIA Management Library) refuses to initialize when this happens.

This almost always means one of two things happened on the host:

1. **An OS package update silently upgraded the NVIDIA driver** (apt/dnf auto-updates, `unattended-upgrades`, etc.) — the new userspace libs are installed, but the kernel is still running the old module. Until the new module is loaded, they mismatch.
2. **You moved to a different host** where the toolkit and driver are out of sync.

The build worked because Docker only needs the toolkit for *running* GPU containers, not for building images. The failure happens at `docker compose up` when nvidia-container-cli tries to inject GPU access into the container.

## Quick diagnostic — confirm the diagnosis

```bash
# 1. Check what the kernel module reports
cat /proc/driver/nvidia/version

# 2. Check what the userspace utility reports
nvidia-smi 2>&1 | head -5

# 3. Check the toolkit version
nvidia-container-cli --version 2>&1 | head -3
```

**If `nvidia-smi` itself errors with the same "driver/library version mismatch"** → it's 100% the kernel-vs-userspace skew. Confirmed.

**If `nvidia-smi` works but containers still fail** → it's the nvidia-container-toolkit specifically. Different fix.

## Fixes, in order of disruption

### Fix 1 (least disruptive) — reload the kernel module

This only works if **no processes are using the GPU**. If your existing stt/tts containers were running, stop them first:

```bash
# Stop everything using the GPU
docker compose down

# Try to remove the NVIDIA modules (in this order)
sudo rmmod nvidia_uvm
sudo rmmod nvidia_drm
sudo rmmod nvidia_modeset
sudo rmmod nvidia

# Reload via modprobe (or just nvidia-smi which auto-loads)
sudo modprobe nvidia
nvidia-smi
```

If `rmmod` fails with `Module nvidia is in use by:` → some process still holds the GPU. Identify it:

```bash
sudo lsof /dev/nvidia*
```

…kill those processes, then retry the `rmmod` sequence. If you can't kill them safely, jump to Fix 3.

### Fix 2 (moderate) — restart the nvidia-container-toolkit + docker

If `nvidia-smi` itself works fine but containers still fail, the toolkit's daemon may be holding stale state:

```bash
sudo systemctl restart nvidia-persistenced  # if installed
sudo systemctl restart docker
```

Then retry `docker compose up -d`.

### Fix 3 (most reliable) — reboot the host

If the host can be rebooted, this is the cleanest fix and takes 2 minutes:

```bash
sudo reboot
```

After reboot:

```bash
nvidia-smi   # should show both T10s cleanly
cd /app/data/voice-stack
docker compose up -d
```

## A partial workaround while you fix the host

If you need to verify *something* works without touching the host, you can bring up only **tts2** — it doesn't use the GPU:

That proves the CPU-only Kokoro service is correctly set up and won't be affected by host GPU issues. The other four services (stt, stt2, tts, tts3) all need the GPU and will fail until the driver issue is fixed.

## Preventing this in the future

The root cause is usually `unattended-upgrades` updating NVIDIA packages on a server that doesn't get rebooted regularly. To prevent silent driver bumps:

```bash
# Pin NVIDIA driver packages so they don't auto-upgrade
sudo apt-mark hold nvidia-driver-* nvidia-utils-* libnvidia-*

# Verify holds
apt-mark showhold | grep nvidia
```

Then upgrade NVIDIA explicitly with a reboot scheduled afterward.

## One bug I hit and fixed

When recreating containers via raw Docker SDK (instead of compose), the service-name DNS aliases (`tts`, `stt2`) didn't get attached automatically. Cloudflared couldn't resolve `tts` → 502 for ~10 min. Fix: reconnected both containers to the network with explicit `aliases=[<container_name>, <service_name>]`. Worth knowing for future SDK-driven recreates — compose adds these implicitly, raw SDK doesn't.

```
WARNING 05-21 20:48:51 [argparse_utils.py:195] With `vllm serve`, you should provide the model as a positional argument or in a config file instead of via the `--model` option. The `--model` option will be removed in v0.13.
(APIServer pid=1) INFO 05-21 20:48:51 [api_server.py:1351] vLLM API server version 0.13.0
(APIServer pid=1) INFO 05-21 20:48:51 [utils.py:253] non-default args: {'model_tag': 'mistralai/Voxtral-Mini-3B-2507', 'host': '0.0.0.0', 'port': 8003, 'api_key': ['66724d1c5d90e4ab26a011b309a8974f9eaf9c06235dbbff2bcd607c17a77fa6'], 'model': 'mistralai/Voxtral-Mini-3B-2507', 'tokenizer_mode': 'mistral', 'dtype': 'float16', 'max_model_len': 8192, 'enforce_eager': True, 'config_format': 'mistral', 'load_format': 'mistral', 'gpu_memory_utilization': 0.85}
(APIServer pid=1) INFO 05-21 20:49:01 [model.py:514] Resolved architecture: VoxtralForConditionalGeneration
(APIServer pid=1) WARNING 05-21 20:49:01 [model.py:2005] Casting torch.bfloat16 to torch.float16.
(APIServer pid=1) INFO 05-21 20:49:01 [model.py:1661] Using max model len 8192
(APIServer pid=1) INFO 05-21 20:49:02 [scheduler.py:230] Chunked prefill is enabled with max_num_batched_tokens=2048.
(APIServer pid=1) WARNING 05-21 20:49:02 [vllm.py:622] Enforce eager set, overriding optimization level to -O0
(APIServer pid=1) INFO 05-21 20:49:02 [vllm.py:722] Cudagraph is disabled under eager mode
(APIServer pid=1) [2026-05-21 20:49:03] INFO tekken.py:204: Non special vocabulary size is 130072 with 1000 special tokens.
(APIServer pid=1) [2026-05-21 20:49:03] INFO tekken.py:532: Cutting non special vocabulary to first 130072 tokens.
(EngineCore_DP0 pid=80) INFO 05-21 20:49:11 [core.py:93] Initializing a V1 LLM engine (v0.13.0) with config: model='mistralai/Voxtral-Mini-3B-2507', speculative_config=None, tokenizer='mistralai/Voxtral-Mini-3B-2507', skip_tokenizer_init=False, tokenizer_mode=mistral, revision=None, tokenizer_revision=None, trust_remote_code=False, dtype=torch.float16, max_seq_len=8192, download_dir=None, load_format=mistral, tensor_parallel_size=1, pipeline_parallel_size=1, data_parallel_size=1, disable_custom_all_reduce=False, quantization=None, enforce_eager=True, kv_cache_dtype=auto, device_config=cuda, structured_outputs_config=StructuredOutputsConfig(backend='auto', disable_fallback=False, disable_any_whitespace=False, disable_additional_properties=False, reasoning_parser='', reasoning_parser_plugin='', enable_in_reasoning=False), observability_config=ObservabilityConfig(show_hidden_metrics_for_version=None, otlp_traces_endpoint=None, collect_detailed_traces=None, kv_cache_metrics=False, kv_cache_metrics_sample=0.01, cudagraph_metrics=False, enable_layerwise_nvtx_tracing=False), seed=0, served_model_name=mistralai/Voxtral-Mini-3B-2507, enable_prefix_caching=True, enable_chunked_prefill=True, pooler_config=None, compilation_config={'level': None, 'mode': <CompilationMode.NONE: 0>, 'debug_dump_path': None, 'cache_dir': '', 'compile_cache_save_format': 'binary', 'backend': 'inductor', 'custom_ops': ['all'], 'splitting_ops': [], 'compile_mm_encoder': False, 'compile_sizes': [], 'compile_ranges_split_points': [2048], 'inductor_compile_config': {'enable_auto_functionalized_v2': False, 'combo_kernels': True, 'benchmark_combo_kernel': True}, 'inductor_passes': {}, 'cudagraph_mode': <CUDAGraphMode.NONE: 0>, 'cudagraph_num_of_warmups': 0, 'cudagraph_capture_sizes': [], 'cudagraph_copy_inputs': False, 'cudagraph_specialize_lora': True, 'use_inductor_graph_partition': False, 'pass_config': {'fuse_norm_quant': False, 'fuse_act_quant': False, 'fuse_attn_quant': False, 'eliminate_noops': False, 'enable_sp': False, 'fuse_gemm_comms': False, 'fuse_allreduce_rms': False}, 'max_cudagraph_capture_size': 0, 'dynamic_shapes_config': {'type': <DynamicShapesType.BACKED: 'backed'>, 'evaluate_guards': False}, 'local_cache_dir': None}
(EngineCore_DP0 pid=80) [2026-05-21 20:49:13] INFO tekken.py:204: Non special vocabulary size is 130072 with 1000 special tokens.
(EngineCore_DP0 pid=80) [2026-05-21 20:49:13] INFO tekken.py:532: Cutting non special vocabulary to first 130072 tokens.
(EngineCore_DP0 pid=80) INFO 05-21 20:49:14 [parallel_state.py:1203] world_size=1 rank=0 local_rank=0 distributed_init_method=tcp://172.18.0.4:53407 backend=nccl
(EngineCore_DP0 pid=80) INFO 05-21 20:49:14 [parallel_state.py:1411] rank 0 in world size 1 is assigned as DP rank 0, PP rank 0, PCP rank 0, TP rank 0, EP rank 0
(EngineCore_DP0 pid=80) ERROR 05-21 20:49:15 [fa_utils.py:73] Cannot use FA version 2 is not supported due to FA2 is only supported on devices with compute capability >= 8
(EngineCore_DP0 pid=80) INFO 05-21 20:49:15 [gpu_model_runner.py:3562] Starting to load model mistralai/Voxtral-Mini-3B-2507...
(EngineCore_DP0 pid=80) WARNING 05-21 20:49:16 [vllm.py:629] Inductor compilation was disabled by user settings,Optimizations settings that are only active duringInductor compilation will be ignored.
(EngineCore_DP0 pid=80) INFO 05-21 20:49:16 [vllm.py:722] Cudagraph is disabled under eager mode
(EngineCore_DP0 pid=80) INFO 05-21 20:49:38 [cuda.py:351] Using FLASHINFER attention backend out of potential backends: ('FLASHINFER', 'TRITON_ATTN', 'FLEX_ATTENTION')
(EngineCore_DP0 pid=80) WARNING 05-21 20:49:38 [vllm.py:629] Inductor compilation was disabled by user settings,Optimizations settings that are only active duringInductor compilation will be ignored.
(EngineCore_DP0 pid=80) INFO 05-21 20:49:38 [vllm.py:722] Cudagraph is disabled under eager mode
(EngineCore_DP0 pid=80) INFO 05-21 20:49:38 [layer.py:477] Using AttentionBackendEnum.TORCH_SDPA for MultiHeadAttention in multimodal encoder.
(EngineCore_DP0 pid=80) INFO 05-21 20:49:40 [weight_utils.py:527] No consolidated.safetensors.index.json found in remote.
Loading safetensors checkpoint shards:   0% Completed | 0/1 [00:00<?, ?it/s]
Loading safetensors checkpoint shards: 100% Completed | 1/1 [00:52<00:00, 52.14s/it]
Loading safetensors checkpoint shards: 100% Completed | 1/1 [00:52<00:00, 52.14s/it]
(EngineCore_DP0 pid=80) 
(EngineCore_DP0 pid=80) INFO 05-21 20:50:32 [default_loader.py:308] Loading weights took 52.32 seconds
(EngineCore_DP0 pid=80) INFO 05-21 20:50:32 [gpu_model_runner.py:3659] Model loading took 8.7223 GiB memory and 76.288070 seconds
(EngineCore_DP0 pid=80) INFO 05-21 20:50:32 [gpu_model_runner.py:4446] Encoder cache will be initialized with a budget of 8192 tokens, and profiled with 1 audio items of the maximum feature size.
(EngineCore_DP0 pid=80) INFO 05-21 20:50:40 [gpu_worker.py:375] Available KV cache memory: 2.34 GiB
(EngineCore_DP0 pid=80) INFO 05-21 20:50:40 [kv_cache_utils.py:1291] GPU KV cache size: 20,416 tokens
(EngineCore_DP0 pid=80) INFO 05-21 20:50:40 [kv_cache_utils.py:1296] Maximum concurrency for 8,192 tokens per request: 2.49x
(EngineCore_DP0 pid=80) INFO 05-21 20:50:40 [kernel_warmup.py:65] Warming up FlashInfer attention.
(EngineCore_DP0 pid=80) INFO 05-21 20:50:43 [core.py:259] init engine (profile, create kv cache, warmup model) took 11.00 seconds
(EngineCore_DP0 pid=80) WARNING 05-21 20:50:43 [vllm.py:629] Inductor compilation was disabled by user settings,Optimizations settings that are only active duringInductor compilation will be ignored.
(EngineCore_DP0 pid=80) INFO 05-21 20:50:43 [vllm.py:722] Cudagraph is disabled under eager mode
(APIServer pid=1) INFO 05-21 20:50:44 [api_server.py:1099] Supported tasks: ['generate', 'transcription']
(APIServer pid=1) INFO 05-21 20:50:46 [api_server.py:1425] Starting vLLM API server 0 on http://0.0.0.0:8003
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:38] Available routes are:
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /openapi.json, Methods: HEAD, GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /docs, Methods: HEAD, GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /docs/oauth2-redirect, Methods: HEAD, GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /redoc, Methods: HEAD, GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /scale_elastic_ep, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /is_scaling_elastic_ep, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /tokenize, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /detokenize, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /inference/v1/generate, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /pause, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /resume, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /is_paused, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /metrics, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /health, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /load, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/models, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /version, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/responses, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/responses/{response_id}, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/responses/{response_id}/cancel, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/messages, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/chat/completions, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/completions, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/audio/transcriptions, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/audio/translations, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /ping, Methods: GET
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /ping, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /invocations, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /classify, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/embeddings, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /score, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/score, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /rerank, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v1/rerank, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /v2/rerank, Methods: POST
(APIServer pid=1) INFO 05-21 20:50:46 [launcher.py:46] Route: /pooling, Methods: POST
(APIServer pid=1) INFO:     Started server process [1]
(APIServer pid=1) INFO:     Waiting for application startup.
(APIServer pid=1) INFO:     Application startup complete.
(APIServer pid=1) INFO:     172.18.0.6:41618 - "GET /ping HTTP/1.1" 200 OK
(APIServer pid=1) INFO:     172.18.0.7:46342 - "GET /health HTTP/1.1" 200 OK
(APIServer pid=1) INFO:     172.18.0.6:41618 - "GET / HTTP/1.1" 404 Not Found
(APIServer pid=1) INFO:     172.18.0.6:41410 - "GET /health HTTP/1.1" 200 OK
(APIServer pid=1) /usr/local/lib/python3.12/dist-packages/mistral_common/tokens/tokenizers/tekken.py:471: FutureWarning: `get_control_token` is deprecated. Use `get_special_token` instead.
(APIServer pid=1)   warnings.warn("`get_control_token` is deprecated. Use `get_special_token` instead.", FutureWarning)
(APIServer pid=1) INFO:     172.18.0.7:57562 - "POST /v1/audio/transcriptions HTTP/1.1" 200 OK
(APIServer pid=1) INFO:     172.18.0.6:60668 - "GET /health HTTP/1.1" 200 OK
(APIServer pid=1) INFO 05-21 20:51:45 [chat_utils.py:590] Detected the chat template content format to be 'string'. You can set `--chat-template-content-format` to override this.
(APIServer pid=1) INFO:     172.18.0.7:60306 - "POST /v1/chat/completions HTTP/1.1" 200 OK
(APIServer pid=1) WARNING 05-21 20:51:46 [processing.py:1153] VoxtralProcessorAdapter did not return `BatchFeature`. Make sure to match the behaviour of `ProcessorMixin` when implementing custom processors.
(APIServer pid=1) INFO:     172.18.0.7:60310 - "POST /v1/chat/completions HTTP/1.1" 200 OK
(APIServer pid=1) INFO:     172.18.0.6:60668 - "GET / HTTP/1.1" 404 Not Found
(APIServer pid=1) INFO 05-21 20:51:53 [loggers.py:248] Engine 000: Avg prompt throughput: 115.6 tokens/s, Avg generation throughput: 9.0 tokens/s, Running: 0 reqs, Waiting: 0 reqs, GPU KV cache usage: 0.0%, Prefix cache hit rate: 31.8%, MM cache hit rate: 33.3%
(APIServer pid=1) INFO 05-21 20:52:03 [loggers.py:248] Engine 000: Avg prompt throughput: 0.0 tokens/s, Avg generation throughput: 0.0 tokens/s, Running: 0 reqs, Waiting: 0 reqs, GPU KV cache usage: 0.0%, Prefix cache hit rate: 31.8%, MM cache hit rate: 33.3%
(APIServer pid=1) INFO:     172.18.0.6:46570 - "GET /health HTTP/1.1" 200 OK
(APIServer pid=1) INFO:     172.18.0.6:53654 - "GET / HTTP/1.1" 404 Not Found
(APIServer pid=1) INFO:     172.18.0.6:45394 - "GET /health HTTP/1.1" 200 OK
(APIServer pid=1) INFO 05-21 20:54:49 [launcher.py:110] Shutting down FastAPI HTTP server.
[rank0]:[W521 20:54:49.178318992 ProcessGroupNCCL.cpp:1524] Warning: WARNING: destroy_process_group() was not called before program exit, which can leak resources. For more info, please see https://pytorch.org/docs/stable/distributed.html#shutdown (function operator())
(APIServer pid=1) INFO:     Shutting down
(APIServer pid=1) INFO:     Waiting for application shutdown.
(APIServer pid=1) INFO:     Application shutdown complete.
 *  Terminal will be reused by tasks, press any key to close it. 

 *  Executing task: docker logs --tail 1000 -f c8089dc58341a744ab8a773137483a7edddc5a948cd73c4a6644e3fae07eb0b3 

WARNING 05-21 20:57:16 [config.py:70] Support for Transformers v4 is deprecated. The Transformers v4 codepath will become unmaintained in vLLM v0.22.0 and will be removed in vLLM v0.24.0. Please upgrade to Transformers v5: pip install --upgrade transformers
WARNING 05-21 20:57:22 [argparse_utils.py:257] With `vllm serve`, you should provide the model as a positional argument or in a config file instead of via the `--model` option. The `--model` option will be removed in a future version.
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306] 
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306]        █     █     █▄   ▄█
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306]  ▄▄ ▄█ █     █     █ ▀▄▀ █  version 0.21.0
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306]   █▄█▀ █     █     █     █  model   mistralai/Voxtral-Mini-3B-2507
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306]    ▀▀  ▀▀▀▀▀ ▀▀▀▀▀ ▀     ▀
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:306] 
(APIServer pid=1) INFO 05-21 20:57:22 [utils.py:240] non-default args: {'model_tag': 'mistralai/Voxtral-Mini-3B-2507', 'host': '0.0.0.0', 'port': 8003, 'api_key': ['66724d1c5d90e4ab26a011b309a8974f9eaf9c06235dbbff2bcd607c17a77fa6'], 'model': 'mistralai/Voxtral-Mini-3B-2507', 'tokenizer_mode': 'mistral', 'dtype': 'float16', 'max_model_len': 8192, 'enforce_eager': True, 'config_format': 'mistral', 'load_format': 'mistral', 'gpu_memory_utilization': 0.85}
(APIServer pid=1) INFO 05-21 20:57:23 [config.py:315] Inferred from consolidated*.safetensors files torch.bfloat16 dtype.
(APIServer pid=1) INFO 05-21 20:57:35 [model.py:568] Resolved architecture: VoxtralForConditionalGeneration
(APIServer pid=1) WARNING 05-21 20:57:35 [model.py:2035] Casting torch.bfloat16 to torch.float16.
(APIServer pid=1) INFO 05-21 20:57:35 [model.py:1697] Using max model len 8192
(APIServer pid=1) INFO 05-21 20:57:35 [vllm.py:886] Asynchronous scheduling is enabled.
(APIServer pid=1) WARNING 05-21 20:57:35 [vllm.py:942] Enforce eager set, disabling torch.compile and CUDAGraphs. This is equivalent to setting -cc.mode=none -cc.cudagraph_mode=none
(APIServer pid=1) WARNING 05-21 20:57:35 [vllm.py:960] Inductor compilation was disabled by user settings, optimizations settings that are only active during inductor compilation will be ignored.
(APIServer pid=1) INFO 05-21 20:57:35 [kernel.py:212] Final IR op priority after setting platform defaults: IrOpPriorityConfig(rms_norm=['vllm_c', 'native'], fused_add_rms_norm=['vllm_c', 'native'])
(APIServer pid=1) INFO 05-21 20:57:35 [vllm.py:1135] Cudagraph is disabled under eager mode
(APIServer pid=1) INFO 05-21 20:57:35 [compilation.py:303] Enabled custom fusions: norm_quant, act_quant
(APIServer pid=1) ERROR 05-21 20:57:38 [fa_utils.py:171] Cannot use FA version 2 is not supported due to FA2 is only supported on devices with compute capability >= 8
WARNING 05-21 20:57:43 [config.py:70] Support for Transformers v4 is deprecated. The Transformers v4 codepath will become unmaintained in vLLM v0.22.0 and will be removed in vLLM v0.24.0. Please upgrade to Transformers v5: pip install --upgrade transformers
(EngineCore pid=115) INFO 05-21 20:57:44 [core.py:109] Initializing a V1 LLM engine (v0.21.0) with config: model='mistralai/Voxtral-Mini-3B-2507', speculative_config=None, tokenizer='mistralai/Voxtral-Mini-3B-2507', skip_tokenizer_init=False, tokenizer_mode=mistral, revision=None, tokenizer_revision=None, trust_remote_code=False, dtype=torch.float16, max_seq_len=8192, download_dir=None, load_format=mistral, tensor_parallel_size=1, pipeline_parallel_size=1, data_parallel_size=1, decode_context_parallel_size=1, dcp_comm_backend=ag_rs, disable_custom_all_reduce=False, quantization=None, quantization_config=None, enforce_eager=True, enable_return_routed_experts=False, kv_cache_dtype=auto, device_config=cuda, structured_outputs_config=StructuredOutputsConfig(backend='auto', disable_any_whitespace=False, disable_additional_properties=False, reasoning_parser='', reasoning_parser_plugin='', enable_in_reasoning=False), observability_config=ObservabilityConfig(show_hidden_metrics_for_version=None, otlp_traces_endpoint=None, collect_detailed_traces=None, kv_cache_metrics=False, kv_cache_metrics_sample=0.01, cudagraph_metrics=False, enable_layerwise_nvtx_tracing=False, enable_mfu_metrics=False, enable_mm_processor_stats=False, enable_logging_iteration_details=False), seed=0, served_model_name=mistralai/Voxtral-Mini-3B-2507, enable_prefix_caching=True, enable_chunked_prefill=True, pooler_config=None, compilation_config={'mode': <CompilationMode.NONE: 0>, 'debug_dump_path': None, 'cache_dir': '', 'compile_cache_save_format': 'binary', 'backend': 'inductor', 'custom_ops': ['all'], 'ir_enable_torch_wrap': False, 'splitting_ops': [], 'compile_mm_encoder': False, 'cudagraph_mm_encoder': False, 'encoder_cudagraph_token_budgets': [], 'encoder_cudagraph_max_vision_items_per_batch': 0, 'encoder_cudagraph_max_frames_per_batch': None, 'compile_sizes': [], 'compile_ranges_endpoints': [2048], 'inductor_compile_config': {'enable_auto_functionalized_v2': False, 'size_asserts': False, 'alignment_asserts': False, 'scalar_asserts': False, 'combo_kernels': True, 'benchmark_combo_kernel': True}, 'inductor_passes': {}, 'cudagraph_mode': <CUDAGraphMode.NONE: 0>, 'cudagraph_num_of_warmups': 0, 'cudagraph_capture_sizes': [], 'cudagraph_copy_inputs': False, 'cudagraph_specialize_lora': True, 'use_inductor_graph_partition': False, 'pass_config': {'fuse_norm_quant': True, 'fuse_act_quant': True, 'fuse_attn_quant': False, 'enable_sp': False, 'fuse_gemm_comms': False, 'fuse_allreduce_rms': False, 'fuse_act_padding': False}, 'max_cudagraph_capture_size': 0, 'dynamic_shapes_config': {'type': <DynamicShapesType.BACKED: 'backed'>, 'evaluate_guards': False, 'assume_32_bit_indexing': False}, 'local_cache_dir': None, 'fast_moe_cold_start': False, 'static_all_moe_layers': []}, kernel_config=KernelConfig(ir_op_priority=IrOpPriorityConfig(rms_norm=['vllm_c', 'native'], fused_add_rms_norm=['vllm_c', 'native']), enable_flashinfer_autotune=False, moe_backend='auto')
(EngineCore pid=115) ERROR 05-21 20:57:45 [fa_utils.py:171] Cannot use FA version 2 is not supported due to FA2 is only supported on devices with compute capability >= 8
(EngineCore pid=115) INFO 05-21 20:57:47 [config.py:315] Inferred from consolidated*.safetensors files torch.bfloat16 dtype.
(EngineCore pid=115) INFO 05-21 20:57:49 [parallel_state.py:1410] world_size=1 rank=0 local_rank=0 distributed_init_method=tcp://172.18.0.3:41403 backend=nccl
(EngineCore pid=115) INFO 05-21 20:57:49 [parallel_state.py:1723] rank 0 in world size 1 is assigned as DP rank 0, PP rank 0, PCP rank 0, TP rank 0, EP rank N/A, EPLB rank N/A
(EngineCore pid=115) Process EngineCore:
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140] EngineCore failed to start.
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140] Traceback (most recent call last):
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 1114, in run_engine_core
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     engine_core = EngineCoreProc(*args, engine_index=dp_rank, **kwargs)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                   ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     return func(*args, **kwargs)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 880, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     super().__init__(
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 118, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.model_executor = executor_class(vllm_config)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                           ^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     return func(*args, **kwargs)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/executor/abstract.py", line 109, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self._init_executor()
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/executor/uniproc_executor.py", line 60, in _init_executor
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.driver_worker.init_device()
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/worker_base.py", line 317, in init_device
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.worker.init_device()  # type: ignore
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     ^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     return func(*args, **kwargs)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/gpu_worker.py", line 330, in init_device
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.model_runner = GPUModelRunnerV1(self.vllm_config, self.device)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                         ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/gpu_model_runner.py", line 488, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.sampler = Sampler(logprobs_mode=self.model_config.logprobs_mode)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                    ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/sample/sampler.py", line 64, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     self.topk_topp_sampler = TopKTopPSampler(logprobs_mode)
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                              ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/sample/ops/topk_topp_sampler.py", line 40, in __init__
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from vllm.v1.attention.backends.flashinfer import FlashInferBackend
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/attention/backends/flashinfer.py", line 11, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from flashinfer import (
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/__init__.py", line 24, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from . import jit as jit
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/__init__.py", line 22, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from . import cubin_loader
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/cubin_loader.py", line 29, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from .core import logger
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/core.py", line 14, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     from . import env as jit_env
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/env.py", line 127, in <module>
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     FLASHINFER_AOT_DIR: pathlib.Path = _get_aot_dir()
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]                                        ^^^^^^^^^^^^^^
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/env.py", line 114, in _get_aot_dir
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140]     raise RuntimeError(
(EngineCore pid=115) ERROR 05-21 20:57:49 [core.py:1140] RuntimeError: flashinfer-jit-cache version (0.5.3+cu129) does not match flashinfer version (0.6.8.post1). Please install the same version of both packages. Set FLASHINFER_DISABLE_VERSION_CHECK=1 to bypass this check.
(EngineCore pid=115) Traceback (most recent call last):
(EngineCore pid=115)   File "/usr/lib/python3.12/multiprocessing/process.py", line 314, in _bootstrap
(EngineCore pid=115)     self.run()
(EngineCore pid=115)   File "/usr/lib/python3.12/multiprocessing/process.py", line 108, in run
(EngineCore pid=115)     self._target(*self._args, **self._kwargs)
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 1144, in run_engine_core
(EngineCore pid=115)     raise e
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 1114, in run_engine_core
(EngineCore pid=115)     engine_core = EngineCoreProc(*args, engine_index=dp_rank, **kwargs)
(EngineCore pid=115)                   ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115)     return func(*args, **kwargs)
(EngineCore pid=115)            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 880, in __init__
(EngineCore pid=115)     super().__init__(
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core.py", line 118, in __init__
(EngineCore pid=115)     self.model_executor = executor_class(vllm_config)
(EngineCore pid=115)                           ^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115)     return func(*args, **kwargs)
(EngineCore pid=115)            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/executor/abstract.py", line 109, in __init__
(EngineCore pid=115)     self._init_executor()
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/executor/uniproc_executor.py", line 60, in _init_executor
(EngineCore pid=115)     self.driver_worker.init_device()
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/worker_base.py", line 317, in init_device
(EngineCore pid=115)     self.worker.init_device()  # type: ignore
(EngineCore pid=115)     ^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(EngineCore pid=115)     return func(*args, **kwargs)
(EngineCore pid=115)            ^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/gpu_worker.py", line 330, in init_device
(EngineCore pid=115)     self.model_runner = GPUModelRunnerV1(self.vllm_config, self.device)
(EngineCore pid=115)                         ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/worker/gpu_model_runner.py", line 488, in __init__
(EngineCore pid=115)     self.sampler = Sampler(logprobs_mode=self.model_config.logprobs_mode)
(EngineCore pid=115)                    ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/sample/sampler.py", line 64, in __init__
(EngineCore pid=115)     self.topk_topp_sampler = TopKTopPSampler(logprobs_mode)
(EngineCore pid=115)                              ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/sample/ops/topk_topp_sampler.py", line 40, in __init__
(EngineCore pid=115)     from vllm.v1.attention.backends.flashinfer import FlashInferBackend
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/attention/backends/flashinfer.py", line 11, in <module>
(EngineCore pid=115)     from flashinfer import (
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/__init__.py", line 24, in <module>
(EngineCore pid=115)     from . import jit as jit
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/__init__.py", line 22, in <module>
(EngineCore pid=115)     from . import cubin_loader
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/cubin_loader.py", line 29, in <module>
(EngineCore pid=115)     from .core import logger
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/core.py", line 14, in <module>
(EngineCore pid=115)     from . import env as jit_env
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/env.py", line 127, in <module>
(EngineCore pid=115)     FLASHINFER_AOT_DIR: pathlib.Path = _get_aot_dir()
(EngineCore pid=115)                                        ^^^^^^^^^^^^^^
(EngineCore pid=115)   File "/usr/local/lib/python3.12/dist-packages/flashinfer/jit/env.py", line 114, in _get_aot_dir
(EngineCore pid=115)     raise RuntimeError(
(EngineCore pid=115) RuntimeError: flashinfer-jit-cache version (0.5.3+cu129) does not match flashinfer version (0.6.8.post1). Please install the same version of both packages. Set FLASHINFER_DISABLE_VERSION_CHECK=1 to bypass this check.
[rank0]:[W521 20:57:49.376971493 ProcessGroupNCCL.cpp:1575] Warning: WARNING: destroy_process_group() was not called before program exit, which can leak resources. For more info, please see https://pytorch.org/docs/stable/distributed.html#shutdown (function operator())
(APIServer pid=1) Traceback (most recent call last):
(APIServer pid=1)   File "/usr/local/bin/vllm", line 7, in <module>
(APIServer pid=1)     sys.exit(main())
(APIServer pid=1)              ^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/cli/main.py", line 92, in main
(APIServer pid=1)     args.dispatch_function(args)
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/cli/serve.py", line 122, in cmd
(APIServer pid=1)     uvloop.run(run_server(args))
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/uvloop/__init__.py", line 96, in run
(APIServer pid=1)     return __asyncio.run(
(APIServer pid=1)            ^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/lib/python3.12/asyncio/runners.py", line 195, in run
(APIServer pid=1)     return runner.run(main)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/lib/python3.12/asyncio/runners.py", line 118, in run
(APIServer pid=1)     return self._loop.run_until_complete(task)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "uvloop/loop.pyx", line 1518, in uvloop.loop.Loop.run_until_complete
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/uvloop/__init__.py", line 48, in wrapper
(APIServer pid=1)     return await main
(APIServer pid=1)            ^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/openai/api_server.py", line 693, in run_server
(APIServer pid=1)     await run_server_worker(listen_address, sock, args, **uvicorn_kwargs)
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/openai/api_server.py", line 707, in run_server_worker
(APIServer pid=1)     async with build_async_engine_client(
(APIServer pid=1)                ^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/lib/python3.12/contextlib.py", line 210, in __aenter__
(APIServer pid=1)     return await anext(self.gen)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/openai/api_server.py", line 100, in build_async_engine_client
(APIServer pid=1)     async with build_async_engine_client_from_engine_args(
(APIServer pid=1)                ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/lib/python3.12/contextlib.py", line 210, in __aenter__
(APIServer pid=1)     return await anext(self.gen)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/entrypoints/openai/api_server.py", line 136, in build_async_engine_client_from_engine_args
(APIServer pid=1)     async_llm = AsyncLLM.from_vllm_config(
(APIServer pid=1)                 ^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/async_llm.py", line 217, in from_vllm_config
(APIServer pid=1)     return cls(
(APIServer pid=1)            ^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/async_llm.py", line 146, in __init__
(APIServer pid=1)     self.engine_core = EngineCoreClient.make_async_mp_client(
(APIServer pid=1)                        ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(APIServer pid=1)     return func(*args, **kwargs)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core_client.py", line 130, in make_async_mp_client
(APIServer pid=1)     return AsyncMPClient(*client_args)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/tracing/otel.py", line 178, in sync_wrapper
(APIServer pid=1)     return func(*args, **kwargs)
(APIServer pid=1)            ^^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core_client.py", line 900, in __init__
(APIServer pid=1)     super().__init__(
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/core_client.py", line 535, in __init__
(APIServer pid=1)     with launch_core_engines(
(APIServer pid=1)          ^^^^^^^^^^^^^^^^^^^^
(APIServer pid=1)   File "/usr/lib/python3.12/contextlib.py", line 144, in __exit__
(APIServer pid=1)     next(self.gen)
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/utils.py", line 1128, in launch_core_engines
(APIServer pid=1)     wait_for_engine_startup(
(APIServer pid=1)   File "/usr/local/lib/python3.12/dist-packages/vllm/v1/engine/utils.py", line 1187, in wait_for_engine_startup
(APIServer pid=1)     raise RuntimeError(
(APIServer pid=1) RuntimeError: Engine core initialization failed. See root cause above. Failed core proc(s): {}
 *  Terminal will be reused by tasks, press any key to close it
 ```
 You can check documentation here.  https://huggingface.co/mistralai/Voxtral-Mini-3B-2507

 IS this the recommended way to create image
 ```
 # stt2 — Voxtral-Mini-3B ASR via vLLM
#
# vLLM ships its own OpenAI-compatible server, so we don't write a main.py.
# The container's job is just to:
#   1. Pin a vLLM base image with the right version
#   2. Add mistral_common[audio] so Voxtral's audio input works
#   3. Let docker-compose pass the launch args at runtime
#
# Why this design:
#   - vLLM exposes /v1/chat/completions (audio attachments → Voxtral does ASR
#     or audio-Q&A through the same endpoint). NOT /v1/audio/transcriptions
#     like our `stt` service. The two stt services have different request
#     shapes — clients pick based on capability.
#   - Voxtral needs `--tokenizer-mode mistral` and friends; passed in compose
#     so they're visible in one place.

FROM vllm/vllm-openai:latest

# Two separate audio extras are needed for Voxtral to actually work:
#
#   - vllm[audio]            — adds librosa + soundfile so vLLM can decode
#                              uploaded audio into tensors. Without it,
#                              /v1/audio/transcriptions and /v1/chat/completions
#                              with audio_url parts fail with HTTP 500
#                              "Please install vllm[audio] for audio support".
#   - mistral_common[audio]  — Mistral's tokenizer extras for chunking
#                              audio tokens. Different layer of the stack.
#
# Both are required. Installing only one leaves audio inference broken.
RUN pip install --no-cache-dir --upgrade \
        "vllm[audio]" \
        "mistral_common[audio]>=1.10.0"

# HuggingFace cache lives on a bind-mounted volume so the ~6 GB Voxtral
# weight download survives container restarts.
ENV HF_HOME=/cache/hf

EXPOSE 8003
```
